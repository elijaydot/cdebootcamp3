# Python Control Flow

## Control Flow

Control flow determines the order in which statements and instructions are executed in a program. In Python, control flow tools like if statements and loops (for and while) allow you to control how your code is executed based on conditions or repetition requirements.

By mastering these control flow statements, you can create dynamic, responsive programs that can handle a wide range of tasks and conditions in Python.



- `If` Statements: Use for decision-making based on conditions. Best for validating input, handling errors, or making choices.
- `for` Loops: Use for iterating over a known sequence of items. Best for repetitive tasks where the number of iterations is known.
- `while` Loops: Use for repeatedly executing code as long as a condition is met. Best when the number of iterations is not known in advance.

## 2. Loops



A loop is a control flow statement that repeatedly executes a block of code as long as a specified condition is `True` or for a specific number of iterations.

Types of Loops in Python
- a. `for` Loop: The `for` loop is used to iterate over a sequence (like a list, tuple, dictionary, set, or string) or any other iterable object.


#### Example:

The `for` loop iterates over each item in the list `fruits`.
For each iteration, it assigns the current item to the variable `fruit` and executes the print statement.

```python
fruits = ['apple', 'banana', 'cherry']

for fruit in fruits:
    print(fruit)
```

#### use cases

b. `while` Loop : The `while` loop is used to repeatedly execute a block of code as long as a specified condition is `True` .



#### Example
The while loop checks if the condition count < 5 is True.
If the condition is True, it executes the code block and then checks the condition again.
The loop continues until the condition becomes False.

```python
count = 0

while count < 5:
    print(count)
    count += 1
```

#### use cases

### `Nested` Loops 

```python
# SELECT count(*) FROM customers where id IS NULL;
   
tables = ['customers', 'orders', 'products', 'prices']
columns = ['id', 'create_date']
for t in tables:
    for c in columns:
        print(f'SELECT count(*) FROM {t} WHERE {c} IS NULL;')
```

### `Break`, `Continue` and `Pass` in control flow
`break` → Immediately exits the nearest loop (stops it completely).

`continue` → Skips the current iteration and moves to the next one.

`pass` → Does nothing; acts as a placeholder when a statement is required but no action is needed.

```python
names = ["Najeeb Arsenal", "Mayor Livepool", " ", "Chief Ilekura", "Joe Manchester", " ", "Musili KWAM1"]
```

```python
for name in names:
    if name == " ":
        print("Empty Name Detected")
        break
    print(f'Full Name = {name}')
```

```python
for name in names:
    if name == " ":
        print("Empty Name Detected")
        continue
    print(f'Full Name = {name}')
```

```python
for name in names:
    if name == " ":
        print("Empty Name Detected")
        pass
    print(f'Full Name = {name}')
```
