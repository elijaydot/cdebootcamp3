## `for` Loops: DE tasks

```python
orders = [
    {"order_id": 1, "customer": "ahmed",  "amount": 5000,  "country": "NG", "status": "paid"},
    {"order_id": 2, "customer": "lahla",  "amount": 0,     "country": "NG", "status": "cancelled"},
    {"order_id": 3, "customer": "judith", "amount": 12000, "country": "GH", "status": "paid"},
    {"order_id": 4, "customer": "ahmed",  "amount": 3000,  "country": "NG", "status": "paid"},
    {"order_id": 5, "customer": "uche",   "amount": 700,   "country": "KE", "status": "pending"},
]

# DE Task 1: count paid orders
paid_count = 0

for order in orders:
    if order["status"] == "paid":
        paid_count += 1

print(f"{paid_count} paid order(s)")
```

```python
# DE Task 2: total of paid orders
paid_total = 0

for order in orders:
    if order["status"] == "paid":
        paid_total = paid_total + order["amount"]

print(f"Paid total: {paid_total}")
```

```python
# DE Task 3: total per country
totals = {}

for order in orders:
    country = order["country"]
    if country in totals:
        totals[country] = totals[country] + order["amount"]
    else:
        totals[country] = order["amount"]

print(totals)
```

```python
# DE Task 4: unique customers
customers = set()

for order in orders:
    customers.add(order["customer"])

print(customers)
print(f"{len(customers)} distinct customer(s)")
```

```python
# DE Task 5: nested loop, one SQL per table + column
tables = ["customers", "orders", "payments"]
columns = ["id", "created_at"]

for table in tables:
    for column in columns:
        print(f"SELECT COUNT(*) FROM {table} WHERE {column} IS NULL;")
```

## `while` Loops: DE tasks

```python
# DE Task 1: polling with a timeout
checks = 0

while checks < 6:
    print("Waiting for file...")
    checks += 1

print(f"Timed out after {checks} checks")
```

```python
# DE Task 2: drain the queue
queue = ["orders", "customers", "payments", "refunds"]

while queue:
    table = queue.pop(0)
    print(f"Loading {table}... {len(queue)} table(s) remaining")

print("All tables loaded")
```

```python
# DE Task 3: batching 47 rows in 10s
total_rows = 47
batch_size = 10
loaded = 0
batch = 0

while loaded < total_rows:
    batch += 1
    rows_left = total_rows - loaded

    if rows_left < batch_size:
        this_batch = rows_left
    else:
        this_batch = batch_size

    loaded += this_batch
    print(f"Batch {batch}: {this_batch} rows (total {loaded}/{total_rows})")

print("Done")
```

```python
# DE Task 4: exponential backoff
wait = 1
attempt = 0
total_wait = 0

while attempt < 5:
    attempt += 1
    print(f"Attempt {attempt}, waiting {wait} second(s)")
    total_wait = total_wait + wait
    wait = wait * 2

print(f"Gave up after {attempt} attempts and {total_wait} seconds of waiting")
```

## `break`, `continue`, `pass`: tasks

```python
# Task 1: stop at the first number over 10
numbers = [3, 8, 12, 5, 20, 7]

for number in numbers:
    if number > 10:
        print(f"Stopped at {number}")
        break
    print(f"Checked {number}")
```

```python
# Task 2: skip blank names
names = ["ahmed", "", "judith", "  ", "uche"]

for name in names:
    if name.strip() == "":
        print("Skipped a blank name")
        continue
    print(f"Name: {name}")
```

```python
# Task 3: count successes until the first failure
logins = ["ok", "ok", "FAILED", "ok"]
successes = 0

for login in logins:
    if login == "FAILED":
        print("Failure detected, stopping the count")
        break
    successes += 1

print(f"{successes} successful login(s) before the failure")
```

```python
# DE Task 4: quarantine with continue
rows = [
    {"id": 1, "email": "a@x.com", "amount": 500},
    {"id": 2, "email": "",        "amount": 900},
    {"id": 3, "email": "c@x.com", "amount": 0},
    {"id": 4, "email": "d@x.com", "amount": 1500},
]

clean = []
quarantine = []

for row in rows:
    if row["email"] == "":
        print(f"Row {row['id']} quarantined: empty email")
        quarantine.append(row)
        continue
    if row["amount"] <= 0:
        print(f"Row {row['id']} quarantined: amount is {row['amount']}")
        quarantine.append(row)
        continue
    clean.append(row)

print()
print(f"Clean: {len(clean)}, Quarantined: {len(quarantine)}")
```

```python
# DE Task 5: abort the whole load with break
clean = []
aborted = False

for row in rows:
    if row["email"] == "" or row["amount"] <= 0:
        print(f"ABORTING LOAD. Row {row['id']} failed validation")
        aborted = True
        break
    clean.append(row)

if aborted:
    clean = []          # nothing gets loaded
    print("Nothing was loaded. Fix the source and rerun")
else:
    print(f"Loaded {len(clean)} row(s)")
```

## Mini project: full worked solution

```python
raw_transactions = [
    {"id": 1, "user": "ahmed",  "amount": 5000,  "currency": "NGN", "status": "success"},
    {"id": 2, "user": "lahla",  "amount": 0,     "currency": "NGN", "status": "failed"},
    {"id": 3, "user": "judith", "amount": 120,   "currency": "USD", "status": "success"},
    {"id": 4, "user": "",       "amount": 800,   "currency": "NGN", "status": "success"},
    {"id": 5, "user": "uche",   "amount": -40,   "currency": "USD", "status": "success"},
    {"id": 6, "user": "ahmed",  "amount": 2500,  "currency": "NGN", "status": "pending"},
    {"id": 7, "user": "judith", "amount": 300,   "currency": "NGN", "status": "success"},
    {"id": 8, "user": "lahla",  "amount": 90,    "currency": "USD", "status": "success"},
]

# Step 1: validate
clean = []
rejected = []
reasons = []

for tx in raw_transactions:
    if tx["user"] == "":
        rejected.append(tx)
        reasons.append(f"id {tx['id']} rejected: empty user")
        continue
    if tx["amount"] <= 0:
        rejected.append(tx)
        reasons.append(f"id {tx['id']} rejected: amount {tx['amount']}")
        continue
    clean.append(tx)

print(f"clean={len(clean)} rejected={len(rejected)}")
```

```python
# Step 2: count by status
status_counts = {}

for tx in clean:
    status = tx["status"]
    if status in status_counts:
        status_counts[status] += 1
    else:
        status_counts[status] = 1

print(status_counts)
```

```python
# Step 3: totals by currency, success only
currency_totals = {}

for tx in clean:
    if tx["status"] != "success":
        continue
    currency = tx["currency"]
    if currency in currency_totals:
        currency_totals[currency] += tx["amount"]
    else:
        currency_totals[currency] = tx["amount"]

print(currency_totals)
```

```python
# Step 4: unique users
users = set()

for tx in clean:
    users.add(tx["user"])

print(users, len(users))
```

```python
# Step 5: biggest transaction
biggest = clean[0]

for tx in clean:
    if tx["amount"] > biggest["amount"]:
        biggest = tx

print(biggest)
```

```python
# Step 6: the report
print("=== PIPELINE REPORT ===")
print(f"Rows in:        {len(raw_transactions)}")
print(f"Rows loaded:    {len(clean)}")
print(f"Rows rejected:  {len(rejected)}")
for reason in reasons:
    print(f"   -> {reason}")

print()
print(f"Status counts:  {status_counts}")
print("Totals (success only):")
for currency, total in currency_totals.items():
    print(f"   {currency}: {total}")

print(f"Distinct users: {len(users)}")
print(f"Largest transaction: id {biggest['id']}, {biggest['user']}, "
      f"{biggest['amount']} {biggest['currency']}")

reject_rate = len(rejected) / len(raw_transactions) * 100
print(f"Rejection rate: {round(reject_rate, 1)}%")
```

```python
# Stretch 7: reject unknown statuses
allowed_statuses = {"success", "failed", "pending"}

for tx in raw_transactions:
    if tx["status"] not in allowed_statuses:
        print(f"id {tx['id']} has an unexpected status: {tx['status']}")

print("Status check done")
```

```python
# Stretch 9: spend per user
spend = {}

for tx in clean:
    user = tx["user"]
    if user in spend:
        spend[user] += tx["amount"]
    else:
        spend[user] = tx["amount"]

for user, amount in spend.items():
    print(f"{user}: {amount}")
```
