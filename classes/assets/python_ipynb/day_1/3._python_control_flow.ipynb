# Python Control Flow

## Control Flow

Control flow determines the order in which statements and instructions are executed in a program. In Python, control flow tools like if statements and loops (for and while) allow you to control how your code is executed based on conditions or repetition requirements.

By mastering these control flow statements, you can create dynamic, responsive programs that can handle a wide range of tasks and conditions in Python.



- `If` Statements: Use for decision-making based on conditions. Best for validating input, handling errors, or making choices.
- `for` Loops: Use for iterating over a known sequence of items. Best for repetitive tasks where the number of iterations is known.
- `while` Loops: Use for repeatedly executing code as long as a condition is met. Best when the number of iterations is not known in advance.

````python
# 1. Even or odd
# % gives you the remainder. If the remainder after dividing by 2 is 0, it is even.

number = 5

if number % 2 == 0:
    print(f"{number} is even")
else:
    print(f"{number} is odd")
````

**Watch out:** `=` assigns a value, `==` compares two values.

````python
# 2. elif chain: cinema ticket price
# Only ONE branch ever runs. Python checks from top to bottom and stops
# at the first condition that is True.

age = 25

if age < 13:
    price = 0
elif age < 18:
    price = 500
elif age < 65:
    price = 1500
else:
    price = 700

print(f"Age {age} pays {price} naira")
````

Try changing `age` to 10, 15, 40 and 70 and run it again.

Order matters. If you put `age < 65` first, a 10 year old would also match it
and pay 1500, because Python stops at the first True.

````python
# 4. Checking a dictionary key before you use it

person = {"name": "Alice", "age": 25}

if "email" in person:
    print(person["email"])
else:
    print("No email on file")
````

Why bother? Because `person["email"]` on its own would crash with a `KeyError`.
The `if` protects you. This is the single most common cause of a pipeline
crashing at 2am: a key that was there yesterday is missing today.

### Real life: `if` in a data pipeline

Before we start, one idea that unlocks everything else:

- **One row of a table is a dictionary.** The column names are the keys.
- **A whole table is a list of those dictionaries.**

That is it. Once you see a table that way, everything you learned about lists
and dicts becomes data engineering.

````python
# One row from a payments table
row = {"customer_id": 101, "email": "dec@gmail.com", "amount": 5000}

print(row["customer_id"])
print(row["email"])
print(row["amount"])
````

````python
# DE example 1: validating a single row before you load it
# This is called a data quality check.

row = {"customer_id": 101, "email": "", "amount": 5000}

if row["email"] == "":
    print(f"REJECT row {row['customer_id']}: email is missing")
elif row["amount"] <= 0:
    print(f"REJECT row {row['customer_id']}: amount must be positive")
else:
    print(f"ACCEPT row {row['customer_id']}")
````

Notice `row['customer_id']` uses single quotes inside the f-string, because the
f-string itself is already using double quotes. Mixing them up is a very common
beginner error.

````python
# DE example 2: choosing how to load a table based on its size

row_count = 1200000

if row_count == 0:
    load_type = "skip, nothing new"
elif row_count < 100000:
    load_type = "full refresh"
else:
    load_type = "incremental"

print(f"{row_count} rows -> {load_type}")
````

````python
# DE example 3: schema check with sets
# The source system promised us 3 columns. Did they all arrive today?

required_columns = {"customer_id", "email", "created_at"}
columns_that_arrived = {"customer_id", "email", "phone", "created_at"}

missing = required_columns - columns_that_arrived   # set difference
extra = columns_that_arrived - required_columns

if missing:
    print(f"PIPELINE FAILED. Missing columns: {missing}")
elif extra:
    print(f"WARNING. New columns appeared: {extra}")
else:
    print("Schema OK")
````

This is called **schema drift** and it breaks real pipelines every week.
Sets make the check one line.

````python
# DE example 4: routing a transaction by its status

transaction = {"id": 9001, "amount": 25000, "status": "failed"}

if transaction["status"] == "success":
    print(f"Send receipt for transaction {transaction['id']}")
elif transaction["status"] == "pending":
    print(f"Hold transaction {transaction['id']} and check again in 5 minutes")
elif transaction["status"] == "failed":
    print(f"Refund transaction {transaction['id']}")
else:
    print(f"Unknown status: {transaction['status']}. Send to the dead letter queue")
````

## Practice: `for` Loops

### Beginner examples

A `for` loop says: "take each item in this thing, one at a time, and do something with it."

```python
for <name_for_one_item> in <the_collection>:
    <do something with that one item>
```

The name after `for` is yours to choose. `for fruit in fruits` reads well.
`for x in fruits` works exactly the same but tells the reader nothing.

````python

# read scores and print

scores = [70, 45, 88, 92, 30]

for score in scores:
    print(score)
````

````python
# 1. The counter pattern
# print only thos who are abpve or equals to 50

scores = [70, 45, 88, 92, 30]


for score in scores:
    if score >= 50:
        print(score)
````

````python
# 1. The counter pattern
# Start at zero, add one every time the condition is met.

scores = [70, 45, 88, 92, 30]
passed = 0

for score in scores:
    if score >= 50:
        #passed = passed + 1
        passed += 1

print(f"{passed} out of {len(scores)} students passed")
````

`passed += 1` is shorthand for `passed = passed + 1`.

The counter has to be created **before** the loop. If you put `passed = 0` inside
the loop it resets to zero on every single item and you always end with 0 or 1.

````python
# 2. The accumulator pattern: adding things up

scores = [70, 45, 88, 92, 30]
total = 0

for score in scores:
    total = total + score

average = total / len(scores)
print(f"Total: {total}")
print(f"Average: {average}")
print(f"Average rounded: {round(average, 1)}")
````

````python
# 3. The filter pattern: building a NEW list from an old one

scores = [70, 45, 88, 92, 30]
passed_scores = []          # start with an empty list

for score in scores:
    if score >= 50:
        passed_scores.append(score)    # .append() adds to the end

print(f"Original: {scores}")
print(f"Passed only: {passed_scores}")
````

These three patterns (count, accumulate, filter) cover most of what you will
ever write in a loop. Learn them well and the rest is detail.

````python
# 4. Looping over a dictionary

prices = {"rice": 1200, "beans": 900, "garri": 500}

print("Keys only:")
for item in prices:
    print(item)

print()
print("Keys and values together:")
for item, price in prices.items():
    print(f"{item} costs {price} naira")
````

````python
# 5. range() gives you numbers to loop over

print("range(3):")
for i in range(3):
    print(i)                 # 0, 1, 2   (starts at 0, stops BEFORE 3)

print()
print("range(1, 4):")
for i in range(1, 4):
    print(i)                 # 1, 2, 3

print()
print("range(0, 10, 2):")
for i in range(0, 10, 2):
    print(i)                 # 0, 2, 4, 6, 8   (the 2 is the step)
````

**The classic mistake:** `range(5)` gives you 0, 1, 2, 3, 4. It does **not** include 5.
If you want 1 to 5, write `range(1, 6)`.

````python
# 6. Bonus: enumerate() when you need the position as well as the item

fruits = ["apple", "banana", "cherry"]

for position, fruit in enumerate(fruits, start=1):
    print(f"{position}. {fruit}")
````

````python
# 7. Looping over a string, and over a tuple

for letter in "data":
    print(letter)

print()

coordinates = (6.5244, 3.3792)     # a tuple
for value in coordinates:
    print(value)
````

### Real life: `for` loops in a data pipeline

Remember: a table is a list of dictionaries. Looping over a table means looping
over that list.

````python
# Our little table for the next few examples
transactions = [
    {"id": 1, "user": "ahmed",  "amount": 5000,  "currency": "NGN", "status": "success"},
    {"id": 2, "user": "lahla",  "amount": 0,     "currency": "NGN", "status": "failed"},
    {"id": 3, "user": "judith", "amount": 120,   "currency": "USD", "status": "success"},
    {"id": 4, "user": "uche",   "amount": 300,   "currency": "NGN", "status": "pending"},
    {"id": 5, "user": "ahmed",  "amount": 80,    "currency": "USD", "status": "success"},
]

# for tx in transactions:
#     print(f"{tx['id']} | {tx['user']:8} | {tx['amount']:>6} {tx['currency']} | {tx['status']}")

for tx in transactions:
    if tx['status'] == 'failed':
        print(tx)
````

`{tx['user']:8}` pads the name to 8 characters so the columns line up.
`{tx['amount']:>6}` pushes the number to the right in 6 characters.
Small thing, but it makes your logs readable.

````python
# DE example 1: counting nulls per column
# This is the first thing you do with any new dataset.

rows = [
    {"id": 1,    "email": "a@x.com", "country": "NG"},
    {"id": 2,    "email": "",        "country": "NG"},
    {"id": None, "email": "c@x.com", "country": ""},
    {"id": 4,    "email": "d@x.com", "country": "GH"},
]

columns = ["id", "email", "country"]

for column in columns:
    nulls = 0
    for row in rows:
        if row[column] is None or row[column] == "":
            nulls += 1
    print(f"{column}: {nulls} null(s) out of {len(rows)} rows")
````

Two new things here.

`is None` is how you check for a real missing value. Use `is None`, not `== None`.

This is a **nested loop**: for every column, walk every row. The inner loop finishes
completely before the outer loop moves to the next column. That is why the counter
`nulls = 0` sits inside the outer loop, so it resets for each column.

````python
# DE example 2: GROUP BY, done by hand
# SELECT currency, SUM(amount) FROM transactions GROUP BY currency

totals = {}

for tx in transactions:
    currency = tx["currency"]
    if currency in totals:
        totals[currency] = totals[currency] + tx["amount"]
    else:
        totals[currency] = tx["amount"]

print(totals)

for currency, total in totals.items():
    print(f"{currency}: {total}")
````

Read that one twice. Every `GROUP BY` you have ever written in SQL is doing
exactly this underneath: walk the rows, and keep a running total per key in a dictionary.

````python
# DE example 3: removing duplicates while keeping the original order

user_ids = [101, 102, 101, 103, 102, 104, 101]

seen = set()
unique_users = []

for uid in user_ids:
    if uid not in seen:
        seen.add(uid)
        unique_users.append(uid)

print(f"Rows in:  {len(user_ids)}")
print(f"Unique:   {len(unique_users)}")
print(f"Result:   {unique_users}")
print(f"Dropped:  {len(user_ids) - len(unique_users)} duplicate row(s)")
````

Why a set and not a list for `seen`? Because checking `in` on a set is fast no
matter how big it gets, while checking `in` on a list gets slower as the list grows.
With 10 million rows that difference is the whole job.

````python
# DE example 4: splitting clean rows from dirty rows
# Real pipelines almost never throw bad rows away. They quarantine them.

clean = []
quarantine = []

for tx in transactions:
    if tx["amount"] <= 0:
        quarantine.append(tx)
    else:
        clean.append(tx)

print(f"Clean rows: {len(clean)}")
print(f"Quarantined: {len(quarantine)}")
for bad in quarantine:
    print(f"  -> id {bad['id']} rejected, amount was {bad['amount']}")
````

## Practice: `while` Loops

### Beginner examples

Use a `for` loop when you know **what** you are looping over.
Use a `while` loop when you only know **when to stop**.

Every safe `while` loop has three parts. Miss one and it runs forever:

1. A starting value, created **before** the loop
2. A condition that will eventually become False
3. Something inside the loop that **changes** the value in the condition

````python
# 1. Countdown

count = 5                 # 1. starting value

while count > 0:          # 2. condition
    print(count)
    count -= 1            # 3. the change. without this line it runs forever

print("Time up!")
````

Delete `count -= 1` in your head and trace it: count stays 5, `5 > 0` is always
True, and the loop never ends. That is an **infinite loop**.

If you ever run one by accident in Jupyter, press the **stop button** (the square)
in the toolbar, or Kernel > Interrupt.

````python
# 2. Attempts, capped at 3

attempts = 0
max_attempts = 3

while attempts < max_attempts:
    attempts += 1
    print(f"Attempt {attempts} of {max_attempts}")

print("No attempts left")
````

````python
# 3. Keep going until you cross a target

balance = 1000
target = 5000
months = 0

while balance < target:
    balance = balance + 800
    months += 1

print(f"Reached {balance} after {months} months")
````

You could not write that as a `for` loop easily, because before running it you
had no idea how many months it would take. That is exactly when `while` earns its place.

````python
# 4. while with a flag variable

logged_in = False
tries = 0

while not logged_in and tries < 3:
    tries += 1
    print(f"Trying to log in... try {tries}")
    if tries == 2:
        logged_in = True

if logged_in:
    print(f"Logged in on try {tries}")
else:
    print("Locked out")
````

### Real life: `while` loops in a data pipeline

`while` is everywhere in data engineering, because so much of the job is
"keep going until the thing is done, and nobody told me how long that will be".

Three patterns cover almost all of it: **retry**, **paginate**, **drain a queue**.

````python
# DE example 1: retrying a flaky connection
# The warehouse is not responding. Try up to 5 times, then give up loudly.

attempt = 0
max_retries = 5
connected = False

while attempt < max_retries and not connected:
    attempt += 1
    print(f"Connecting to warehouse... attempt {attempt}")
    if attempt == 3:          # pretend the 3rd attempt succeeds
        connected = True

if connected:
    print(f"Connected after {attempt} attempt(s)")
else:
    print("Could not connect after 5 tries. Page the on-call engineer")
````

Notice the `and not connected` in the condition. Without it, the loop would keep
retrying all 5 times even after it succeeded. Two things must be true to continue:
we still have retries left, **and** we are not connected yet.

````python
# DE example 2: pagination
# An API gives you 2 rows at a time. You do not know how many pages exist,
# so you keep asking until it hands you an empty page.

pages = [
    ["row1", "row2"],
    ["row3", "row4"],
    ["row5"],
    [],                    # the API says "nothing left"
]

page_number = 0
all_rows = []

while page_number < len(pages):
    page = pages[page_number]

    if not page:                       # empty page means we are done
        print("Empty page received. Finished fetching")
        break

    print(f"Fetched page {page_number + 1} with {len(page)} row(s)")
    for row in page:
        all_rows.append(row)

    page_number += 1

print(f"Total rows pulled: {len(all_rows)}")
print(all_rows)
````

````python
# DE example 3: draining a queue of files
# Keep processing while the list still has something in it.

queue = ["sales_jan.csv", "sales_feb.csv", "sales_mar.csv"]

while queue:
    current_file = queue.pop(0)        # .pop(0) removes and returns the FIRST item
    print(f"Processing {current_file}. {len(queue)} file(s) left in the queue")

print("Queue is empty. Pipeline done")
````

`while queue:` works because an empty list is False. The loop stops on its own
the moment the last file is popped. And `.pop(0)` is what makes the queue shrink,
so this loop can never run forever.

````python
# DE example 4: batching a big load into chunks

total_rows = 23
batch_size = 5
loaded = 0
batch_number = 0

while loaded < total_rows:
    batch_number += 1
    rows_left = total_rows - loaded

    if rows_left < batch_size:
        this_batch = rows_left         # the last batch is smaller
    else:
        this_batch = batch_size

    loaded += this_batch
    print(f"Batch {batch_number}: loaded {this_batch} rows. Running total {loaded}/{total_rows}")

print("Load complete")
````

`continue` → Skips the current iteration and moves to the next one.

## Practice: `break`, `continue` and `pass`

### The one-line summary

| Keyword | What it does | When you reach for it |
|---|---|---|
| `break` | Leaves the loop completely, right now | You found what you wanted, or something is so wrong you must stop |
| `continue` | Skips the rest of this round, goes to the next item | This one item is bad, but the others deserve a chance |
| `pass` | Nothing at all. A placeholder | You need a line there but have not decided the logic yet |

### A bug worth staring at

Look carefully at this loop. The `print` never runs. Not once.

```python
for email in emails:
    if ';' in email:
        continue
        print('SQL Injection: Hacker Attack')     # dead code
    print(f'Processing Email: {email}')
```

`continue` jumps straight to the next item. Any line you write **after** it inside
that same block is unreachable. Same with `break`. Python will not warn you.

Put your `print` (or your logging, or your quarantine append) **before** the
`continue`, not after.

````python
# Fixed version: say something, THEN skip

emails = [
    "dec@gmail.com",
    "cde@outlook.de",
    "DROP TABLE USERS;",
    "correctguy@gmail.com",
    "lay@gmail.com",
]

for email in emails:
    if ";" in email:
        print(f"REJECTED (suspicious): {email}")     # first
        continue                                     # then skip
    print(f"Processing email: {email}")
````

### Fail fast, or skip and carry on?

This is a real decision you make on every pipeline, and there is no single right answer.

- **Fail fast (`break`)**: one bad row means the whole batch is untrustworthy.
  Money movement, ledger entries, anything where a half-loaded batch is worse than
  no batch at all.
- **Skip and carry on (`continue`)**: one bad row is just one bad row. Log it, set it
  aside, keep the good 99%. Clickstream, sensor readings, marketing events.

The mistake is not picking one. The mistake is skipping bad rows **silently**, so
nobody ever finds out they went missing.

````python
# Fail fast with break

rows = [
    {"id": 1, "amount": 100},
    {"id": 2, "amount": -50},
    {"id": 3, "amount": 300},
]

for row in rows:
    if row["amount"] < 0:
        print(f"STOPPING THE LOAD. Bad row found: {row}")
        break
    print(f"Loaded row {row['id']}")

print("Loop finished")
````

````python
# Skip and quarantine with continue

loaded = []
rejected = []

for row in rows:
    if row["amount"] < 0:
        rejected.append(row)
        continue
    loaded.append(row)

print(f"Loaded:   {len(loaded)} row(s)")
print(f"Rejected: {len(rejected)} row(s)")
for bad in rejected:
    print(f"  -> id {bad['id']}, amount {bad['amount']}")
````

````python
# break also means "stop as soon as you find it"
# No point scanning 10 million rows after you already have your answer.

user_ids = [101, 102, 103, 104, 105]
looking_for = 103
found = False

for uid in user_ids:
    print(f"checking {uid}")
    if uid == looking_for:
        found = True
        break

if found:
    print(f"Found {looking_for}. Stopped early, did not scan the rest")
else:
    print(f"{looking_for} is not in the list")
````

`pass` is not a data engineering tool. It is a "I am still thinking" tool.
It lets you write the shape of your logic before you know the details, without
Python throwing an `IndentationError` at an empty block.

If a `pass` is still in your code at review time, either fill it in or delete the branch.

## Common mistakes to watch for

| Mistake | What you see | The fix |
|---|---|---|
| `if x = 5:` | `SyntaxError` | Use `==` to compare. `=` only assigns |
| Forgot the colon | `SyntaxError: expected ':'` | Every `if`, `elif`, `else`, `for`, `while` line ends in `:` |
| Wrong indentation | `IndentationError` | Everything inside a block is indented the same amount. Pick 4 spaces and never mix tabs and spaces |
| Counter inside the loop | Your total is always 0 or 1 | `total = 0` goes **before** the loop, not inside it |
| `while` that never ends | Notebook hangs, `[*]` never finishes | Something inside the loop must change the condition variable. Press the stop button to escape |
| Code after `continue` or `break` | It silently never runs | Move it above the `continue` / `break` |
| `range(5)` when you wanted 1 to 5 | You get 0,1,2,3,4 | `range(1, 6)` |
| `row["email"]` on a missing key | `KeyError` | Check `if "email" in row:` first |
| `if value == None` | Works, but wrong style | Use `if value is None` |
| Changing a list while looping over it | Items get skipped | Build a new list instead, or loop over a copy |

````python
# See the counter mistake for yourself. This one is WRONG.

scores = [70, 45, 88]

for score in scores:
    total = 0              # resets on every single item
    total = total + score

print(f"Wrong answer: {total}")

# And this one is right.

total = 0                  # created once, before the loop
for score in scores:
    total = total + score

print(f"Right answer: {total}")
````
