## Tuples
Tuples are ordered, immutable, and allow duplicate elements. Since they are immutable, their contents cannot be modified after creation.
### Tuple Operations

```python
A tuple:
- Is ordered
- Cannot be changed (immutable)
- Allows duplicates
```

```python
# Create a tuple
my_tuple = (1, 2, 3, 4, 5)

# Access elements by index
print(my_tuple[0])      # First element
print(my_tuple[-1])     # Last element

# Slicing
print(my_tuple[1:4])    # Slice from index 1 to 3
```

### Common Tuple Methods

```python
# Tuples have fewer methods than lists due to their immutability.
# Two common methods are `count` and `index`.

# Count: Count the number of occurrences of a value
print("Count of 3:", my_tuple.count(3))

# Index: Find the index of the first occurrence of a value
print("Index of 4:", my_tuple.index(4))
```

## Sets
Sets are unordered collections of unique elements. They are used to perform operations like union, intersection, and difference.
### Set Operations and Methods

```python
A set:
- Is unordered
- Cannot have duplicates
- Is changeable
```

```python
my_set = {1, 2, 3, 4, 5}
```

```python
my_set
```

```python
# Create a set
my_set = {1, 2, 3, 4, 5}

# Add: Add an element to the set
my_set.add(6)
print(my_set)

# Remove: Remove an element (raises KeyError if element is not found)
my_set.remove(3)
print(my_set)

# Discard: Remove an element (does not raise an error if element is not found)
my_set.discard(16)   # No error
print(my_set)

# Pop: Remove and return a random element
random_item = my_set.pop()
print("Popped item:", random_item)
print(my_set)

# Clear: Remove all elements from the set
my_set.clear()
print("Cleared set:", my_set)
```

### Set Operations

```python
set_a = {1, 2, 3}
set_b = {3, 4, 5}

# Union: Combine two sets (all unique elements)
print("Union:", set_a.union(set_b))

# Intersection: Elements present in both sets
print("Intersection:", set_a.intersection(set_b))

# Difference: Elements in set_a but not in set_b
print("Difference:", set_a.difference(set_b))

# Symmetric Difference: Elements in either set, but not both
print("Symmetric Difference:", set_a.symmetric_difference(set_b))
```

## Dictionaries
Dictionaries are collections of key-value pairs. They are unordered, mutable, and allow for fast lookups by key.
### Dictionary Operations and Methods

```python
Dictionary:
- Stores data as key-value pairs
- Is ordered (Python 3.7+)
- Is changeable
```

```python
# Create a dictionary
my_dict = {'a': 1, 'b': 2, 'c': 3}

# Access a value by key
print(my_dict['a'])

# Add or update a key-value pair
my_dict['d'] = 4
print(my_dict)

# Remove a key-value pair
removed_value = my_dict.pop('b')
print("Removed value:", removed_value)
print(my_dict)

# Get: Access a value safely (returns None if key is not found)
value = my_dict.get('e', 'Not found')
print("Value for key 'e':", value)

# Keys, values, and items
print("Keys:", my_dict.keys())
print("Values:", my_dict.values())
print("Items:", my_dict.items())
```
