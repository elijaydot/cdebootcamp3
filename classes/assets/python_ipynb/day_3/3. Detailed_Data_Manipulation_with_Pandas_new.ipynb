### Series and DataFrame: two objects, not one

A **Series** is one column. A **DataFrame** is a collection of Series sharing an index.
Almost every confusing pandas error comes from not knowing which one you are holding.

````python
import pandas as pd

one_column = df["Month"]              # single brackets -> Series
two_columns = df[["Month"]]           # double brackets -> DataFrame with one column

print(type(one_column).__name__)
print(type(two_columns).__name__)
print()
print(one_column.head(3))
print()
print(two_columns.head(3))
````

Single brackets give you a Series. Double brackets give you a DataFrame.
`df["a"]` and `df[["a"]]` look almost identical and behave differently, and that is
worth ten minutes of confusion now instead of an hour later.

````python
# A Series has an index and values
s = df["Month"]

print(f"name:   {s.name}")
print(f"dtype:  {s.dtype}")
print(f"shape:  {s.shape}")
print(f"index:  {list(s.index)[:5]}")
print(f"values: {s.values[:5]}")
````

## Practice: Selecting and filtering

### `.loc` and `.iloc`: the two you must not mix up

- `.loc[rows, columns]` selects by **label**. The stop is **included**.
- `.iloc[rows, columns]` selects by **position**. The stop is **excluded**, like a list.

If you only remember one thing: `loc` = label, `iloc` = index position.

````python
import pandas as pd
orders = pd.read_csv("orders.csv", parse_dates=["order_date"])

print("iloc[0] gives the first row as a Series:")
print(orders.iloc[0])
print()
print("iloc[0:3] gives 3 rows. 0, 1, 2. Not 3.")
print(orders.iloc[0:3][["order_id", "item"]])
````

````python
print("loc[0:3] gives FOUR rows, because loc includes the stop label:")
print(orders.loc[0:3][["order_id", "item"]])
````

That difference bites everyone once. `iloc[0:3]` gives 3 rows, `loc[0:3]` gives 4.
`loc` is inclusive because labels are not necessarily numbers, so "up to and
including this label" is the only definition that makes sense.

````python
# Rows and columns together
print(orders.loc[0:2, ["order_id", "item", "unit_price"]])
print()
print(orders.iloc[0:2, 0:3])          # first 2 rows, first 3 columns
print()
print(orders.loc[5, "item"])          # one cell, by label
print(orders.iloc[5, 2])              # the same cell, by position
````

````python
# Filtering: a boolean mask is just a Series of True and False
mask = orders["unit_price"] > 100000

print(mask.head(8))
print()
print(f"{mask.sum()} rows match")       # True counts as 1
print()
print(orders[mask][["order_id", "item", "unit_price"]])
````

### The parentheses rule, which trips up literally everyone

Combining conditions in pandas uses `&`, `|` and `~`, **not** `and`, `or`, `not`.
And every condition must be wrapped in its own parentheses, because `&` binds more
tightly than `>` does.

````python
# This is correct
result = orders[(orders["unit_price"] > 50000) & (orders["country"] == "NG")]
print(result[["order_id", "item", "unit_price", "country"]])
````

````python
# Forget the parentheses and you get a confusing error, not a wrong answer
try:
    orders[orders["unit_price"] > 50000 & orders["country"] == "NG"]
except Exception as e:
    print(f"{type(e).__name__}: {e}")

print()
# Use "and" instead of "&" and you get a different confusing error
try:
    orders[(orders["unit_price"] > 50000) and (orders["country"] == "NG")]
except Exception as e:
    print(f"{type(e).__name__}: {str(e)[:120]}")
````

"The truth value of a Series is ambiguous" is pandas saying: you gave me a whole
column of True and False and asked me to decide if it is one True or one False.
Use `&` for element-by-element AND, `|` for OR, `~` for NOT.

````python
# The three operators in use
electronics_ng = orders[(orders["category"] == "Electronics") & (orders["country"] == "NG")]
gh_or_ke = orders[(orders["country"] == "GH") | (orders["country"] == "KE")]
not_completed = orders[~(orders["status"] == "completed")]

print(f"electronics in NG: {len(electronics_ng)}")
print(f"GH or KE:          {len(gh_or_ke)}")
print(f"not completed:     {len(not_completed)}")
````

````python
# isin, between and str: shorter than chaining ORs
print(orders[orders["country"].isin(["GH", "KE"])][["order_id", "country"]])
print()
print(orders[orders["unit_price"].between(10000, 100000)][["order_id", "unit_price"]])
print()
print(orders[orders["item"].str.contains("Laptop", na=False)][["order_id", "item"]])
print()
print(orders[orders["status"] != "completed"][["order_id", "status"]])
````

`na=False` on `.str.contains` matters. Without it, a null in the column produces
`NaN` rather than `False`, and pandas refuses to use that as a mask.

````python
# query(): the same filter, written like SQL. Easier to read for long conditions.
print(orders.query("unit_price > 50000 and country == 'NG'")[["order_id", "item", "unit_price"]])
print()
threshold = 100000
print(orders.query("unit_price > @threshold")[["order_id", "unit_price"]])
````

Inside `query`, `@threshold` means "the Python variable called threshold".
And note that `query` does use the words `and` / `or`, unlike bracket filtering.
Yes, that is inconsistent. Yes, it catches people out.

### `SettingWithCopyWarning`, and how to never see it again

This is the most common pandas warning in the world, and the fix is one word.

````python
import warnings

# The pattern that causes it: filter, then assign to the result
subset = orders[orders["country"] == "NG"]

with warnings.catch_warnings(record=True) as caught:
    warnings.simplefilter("always")
    subset["discounted"] = subset["unit_price"] * 0.9
    if caught:
        print(f"WARNING: {caught[0].category.__name__}")
    else:
        print("no warning on this pandas version, but the habit below is still right")

print(subset[["order_id", "unit_price", "discounted"]].head(3))
````

The problem is that `orders[...]` may hand you a **view** onto the original rather
than a new DataFrame, so pandas cannot tell whether you meant to change the subset
or the original. Behaviour differs between pandas versions, which is worse than a
clear error.

The fix, in both pandas 2 and pandas 3: **add `.copy()` when you intend a new,
independent table.**

````python
subset = orders[orders["country"] == "NG"].copy()      # explicit, no ambiguity
subset["discounted"] = subset["unit_price"] * 0.9

print(subset[["order_id", "unit_price", "discounted"]].head(3))
print()
print(f"original untouched? {'discounted' not in orders.columns}")
````

````python
# When you DO want to change the original, use .loc on the original
orders_copy = orders.copy()
orders_copy.loc[orders_copy["country"] == "NG", "region"] = "West Africa"
orders_copy.loc[orders_copy["country"] == "KE", "region"] = "East Africa"
orders_copy.loc[orders_copy["country"] == "GH", "region"] = "West Africa"

print(orders_copy[["order_id", "country", "region"]].head(6))
````

**The rule:** to change the original, `df.loc[mask, "column"] = value`.
To work on a subset, `df[mask].copy()` first. Never assign into a filtered result
you did not copy.

## Handling Missing Data
Pandas provides easy ways to detect, fill, or drop missing values.

````python
df.dropna() 
````

**Reference: missing data**

```python
df.isna().sum()                      # count of nulls per column
df.isna().mean() * 100               # null rate as a percentage

df.dropna()                          # drop rows with ANY null. usually too aggressive
df.dropna(subset=["amount"])         # drop only where this column is null
df.dropna(axis=1)                    # drop COLUMNS containing nulls

df.fillna(0)                         # replace nulls with 0
df.fillna(df.mean(numeric_only=True))  # column means. numeric_only is required
```

`df.fillna(df.mean())` without `numeric_only=True` raises a `TypeError` on any table
that has a text column. Worked examples are in the practice section below.

## Practice: Missing data and data types

### First, see what is missing

````python
import pandas as pd
orders = pd.read_csv("orders.csv", parse_dates=["order_date"])

print(orders.isna().sum())
print()
print("rows with ANY null:")
print(orders[orders.isna().any(axis=1)])
````

`axis=1` means "look across the row". `axis=0` means "look down the column".
Whenever a pandas function takes `axis`, that is what it is asking.

````python
# Three ways to fill, and they are not interchangeable
filled_zero = orders["unit_price"].fillna(0)
filled_mean = orders["unit_price"].fillna(orders["unit_price"].mean())
filled_ffill = orders["unit_price"].ffill()          # carry the previous value forward

comparison = pd.DataFrame({
    "original": orders["unit_price"],
    "fill_0": filled_zero,
    "fill_mean": filled_mean.round(0),
    "ffill": filled_ffill,
})
comparison
````

Choosing wrongly here quietly corrupts a report. Filling a missing price with `0`
makes your revenue too low. Filling with the mean invents money that never existed.
For a price, the honest answer is usually to leave it null and exclude the row from
the revenue calculation, saying so in the output.

`ffill` belongs to time series, where carrying the last known reading forward is
meaningful. It is the wrong default for anything else.

````python
# df.fillna(df.mean()) fails on a table with text columns. This is the fix.
try:
    orders.fillna(orders.mean())
except TypeError as e:
    print(f"TypeError: {str(e)[:90]}")

print()
numeric_means = orders.mean(numeric_only=True)
print(numeric_means)
print()
filled = orders.fillna(numeric_means)
print(f"nulls before: {orders.isna().sum().sum()}, after: {filled.isna().sum().sum()}")
````

````python
# dropna: be specific about what you are dropping
print(f"all rows:                      {len(orders)}")
print(f"dropna() any null anywhere:    {len(orders.dropna())}")
print(f"dropna(subset=['unit_price']): {len(orders.dropna(subset=['unit_price']))}")
print(f"dropna(how='all'):             {len(orders.dropna(how='all'))}")
print(f"dropna(axis=1) drops COLUMNS:  {orders.dropna(axis=1).shape}")
````

A bare `dropna()` drops a row if **any** column is null. On a wide table that can
delete most of your data because of one mostly-empty column nobody uses.
Always pass `subset=` and name the columns that actually matter.

### Types: where silent wrongness lives

````python
print(orders.dtypes)
print()
print(f"unit_price is numeric:  {pd.api.types.is_numeric_dtype(orders['unit_price'])}")
print(f"order_date is a date:   {pd.api.types.is_datetime64_any_dtype(orders['order_date'])}")
````

````python
# The classic disaster: a numeric column read as text because of one bad value
messy = pd.DataFrame({"amount": ["1000", "2500", "N/A", "750", ""]})
print(messy.dtypes)

# Summing it does something, and that something is wrong
print(f"sum of a text column: {messy['amount'].sum()!r}")
````

````python
# to_numeric with errors="coerce" turns anything unparseable into NaN
messy["amount_clean"] = pd.to_numeric(messy["amount"], errors="coerce")
print(messy)
print()
print(f"dtype now: {messy['amount_clean'].dtype}")
print(f"sum:       {messy['amount_clean'].sum()}")
print(f"coerced to null: {messy['amount_clean'].isna().sum()} value(s)")
````

`errors="coerce"` is the single most useful argument in pandas cleaning. It says
"convert what you can, make the rest null, do not crash". Then you decide what to do
with the nulls, deliberately, instead of the job dying on row 4,000,000.

Always report how many values it coerced. A silent coercion is a silent data loss.

````python
# astype vs to_numeric: astype is strict, to_numeric is forgiving
try:
    messy["amount"].astype(float)
except ValueError as e:
    print(f"astype raises: {str(e)[:70]}")

print("to_numeric coerces instead:", list(pd.to_numeric(messy["amount"], errors="coerce")))
````

````python
# Dates: the same story
raw_dates = pd.Series(["2026-01-15", "15/01/2026", "not a date", None])

print(pd.to_datetime(raw_dates, errors="coerce", format="mixed", dayfirst=False))
print()
print("with an explicit format, anything that does not match becomes NaT:")
print(pd.to_datetime(raw_dates, errors="coerce", format="%Y-%m-%d"))
````

`NaT` is "not a time", the date version of `NaN`. Specifying `format=` explicitly is

## Practice: Grouping and aggregating

`groupby` is `GROUP BY`. If you can write the SQL, you can write the pandas.

```sql
SELECT country, SUM(revenue) FROM orders GROUP BY country
```
becomes
```python
orders.groupby("country")["revenue"].sum()
```

````python
import pandas as pd
orders = pd.read_csv("orders.csv", parse_dates=["order_date"])
orders["revenue"] = orders["quantity"] * orders["unit_price"]

print(orders.groupby("country")["revenue"].sum())
print()
print(orders.groupby("country")["revenue"].sum().sort_values(ascending=False))
````

That returns a **Series**, indexed by country. Add `.reset_index()` to get a normal
DataFrame back, which is usually what you want for exporting or merging.

````python
by_country = orders.groupby("country")["revenue"].sum().reset_index()
print(type(by_country).__name__)
by_country
````

````python
# Several aggregations at once
print(orders.groupby("country")["revenue"].agg(["count", "sum", "mean", "min", "max"]))
````

````python
# Different aggregations for different columns
result = orders.groupby("country").agg({
    "revenue": "sum",
    "quantity": "sum",
    "order_id": "count",
    "unit_price": "mean",
})
result
````

````python
# Named aggregation: the version you will actually want, because the columns get real names
summary = orders.groupby("country").agg(
    orders=("order_id", "count"),
    customers=("customer_id", "nunique"),
    total_revenue=("revenue", "sum"),
    avg_order_value=("revenue", "mean"),
    biggest_order=("revenue", "max"),
).reset_index()

summary["avg_order_value"] = summary["avg_order_value"].round(0)
summary.sort_values("total_revenue", ascending=False)
````

Compare the column names here with the ones from `agg(["count", "sum"])`. Named
aggregation produces a table you can hand to someone without explaining it.

````python
# Grouping by more than one column
multi = orders.groupby(["country", "category"]).agg(
    orders=("order_id", "count"),
    revenue=("revenue", "sum"),
).reset_index()

multi.sort_values("revenue", ascending=False)
````

````python
# Grouping by something you compute on the fly
by_month = orders.groupby(orders["order_date"].dt.strftime("%Y-%m")).agg(
    orders=("order_id", "count"),
    revenue=("revenue", "sum"),
).reset_index(names="month")

by_month
````

The null `order_date` on order 1011 simply does not appear. `groupby` drops null
keys by default. Pass `dropna=False` if a missing key is itself meaningful, which in
a data quality report it usually is.

````python
print("default, nulls dropped:")
print(orders.groupby("status", dropna=True)["order_id"].count())
print()
print("with dropna=False, you can see them:")
print(orders.groupby(orders["order_date"].dt.year, dropna=False)["order_id"].count())
````

````python
# transform: an aggregate broadcast back onto every row.
# This is the pandas answer to a SQL window function.
orders["country_total"] = orders.groupby("country")["revenue"].transform("sum")
orders["share_of_country"] = (orders["revenue"] / orders["country_total"] * 100).round(1)

orders[["order_id", "country", "revenue", "country_total", "share_of_country"]].head(8)
````

`transform` keeps the original number of rows, `agg` collapses them. That is the
whole difference, and it is exactly `SUM(...) OVER (PARTITION BY country)` versus
`GROUP BY country`.

````python
# pivot_table: a group-by with one of the keys turned into columns
pivot = orders.pivot_table(
    values="revenue",
    index="country",
    columns="category",
    aggfunc="sum",
    fill_value=0,
    margins=True,               # adds a total row and column
)
pivot
````

You can combine multiple DataFrames using merge or join operations.

```python
merged = pd.merge(df1, df2, on="common_column", how="inner")
merged = pd.merge(df1, df2, on="common_column", how="left")

)
```

---

## Practice: Merging, joining and concatenating

`merge` joins **across**, adding columns. `concat` stacks **down**, adding rows.
Mixing these up is the most common beginner mistake in this section.

````python
import pandas as pd

orders = pd.read_csv("orders.csv", parse_dates=["order_date"])
customers = pd.read_csv("customers.csv", parse_dates=["signup_date"])

print(f"orders:    {orders.shape}, customer_ids: {sorted(orders['customer_id'].unique())}")
print(f"customers: {customers.shape}, customer_ids: {sorted(customers['customer_id'].unique())}")
print()
print("note: orders has customer 106 and 999, customers has 107. They do not match perfectly.")
````

````python
# The four join types, on data that deliberately does not line up
for how in ["inner", "left", "right", "outer"]:
    merged = orders.merge(customers, on="customer_id", how=how)
    print(f"{how:>6}: {len(merged)} rows, {merged['name'].isna().sum()} null names")
````

| `how` | Keeps | Use it when |
|---|---|---|
| `inner` | Only matching rows from both | You need complete records only |
| `left` | Every row from the left, nulls where no match | **The default in DE.** Keep all facts, enrich where possible |
| `right` | Every row from the right | Rare. Just swap the tables and use left |
| `outer` | Everything from both | Reconciliation, when you want to see both kinds of orphan |

`left` is the workhorse. You almost never want to silently lose fact rows, which is
what `inner` does.

````python
# indicator=True tells you where every row came from. Use it every time you investigate.
merged = orders.merge(customers, on="customer_id", how="outer", indicator=True)

print(merged["_merge"].value_counts())
print()
print("orders with no matching customer (orphan facts):")
print(merged[merged["_merge"] == "left_only"][["order_id", "customer_id", "item"]])
print()
print("customers with no orders (never bought anything):")
print(merged[merged["_merge"] == "right_only"][["customer_id", "name", "segment"]])
````

Those orphans are a real finding, not noise. Customer 999 on order 1012 does not
exist in the customer table, which usually means the two tables were extracted at
different times, or someone deleted a customer without cleaning up. Customer 107
signed up and never ordered, which is a marketing question rather than a bug.

````python
# validate= makes pandas check your assumption about the relationship
ok = orders.merge(customers, on="customer_id", how="left", validate="many_to_one")
print(f"many_to_one passed: {len(ok)} rows")

# Now break it: a duplicated customer row would silently multiply your revenue
broken_customers = pd.concat([customers, customers.iloc[[0]]], ignore_index=True)

try:
    orders.merge(broken_customers, on="customer_id", how="left", validate="many_to_one")
except Exception as e:
    print(f"caught: {type(e).__name__}: {str(e)[:80]}")

no_check = orders.merge(broken_customers, on="customer_id", how="left")
print(f"without validate: {len(orders)} orders became {len(no_check)} rows. Revenue now double counts.")
````

**Use `validate=` on every merge.** A duplicated key on the right-hand side silently
multiplies your rows, and the only symptom is that a number in a report is too big.
Nobody notices for three months. `validate="many_to_one"` turns that into an
exception the moment it happens.

````python
# Different column names on each side
left = orders.rename(columns={"customer_id": "cust"})
merged = left.merge(customers, left_on="cust", right_on="customer_id", how="left")
print(merged[["order_id", "cust", "name"]].head(3))
````

````python
# Overlapping column names get suffixed. Control the suffixes so the result is readable.
merged = orders.merge(customers, on="customer_id", how="left", suffixes=("_order", "_customer"))
print([c for c in merged.columns if "country" in c])
print()
print(merged[["order_id", "country_order", "country_customer"]].head(5))
````

Both tables have a `country` column, so pandas kept both and suffixed them. That is
also a free data quality check: if the order country and the customer country
disagree, something is worth looking at.

````python
mismatch = merged[
    merged["country_order"].notna()
    & merged["country_customer"].notna()
    & (merged["country_order"] != merged["country_customer"])
]
print(f"{len(mismatch)} order(s) where the order country differs from the customer country")
print(mismatch[["order_id", "name", "country_order", "country_customer"]])
````

````python
# concat: stacking two months of the same table
may = pd.read_csv("orders_may.csv", parse_dates=["order_date"])

stacked = pd.concat([orders, may], ignore_index=True)
print(f"{len(orders)} + {len(may)} = {len(stacked)} rows")
print(stacked.tail(4)[["order_id", "item", "order_date"]])
````

`ignore_index=True` renumbers the index. Without it you get two rows numbered 0,
two numbered 1, and every later `.loc` lookup becomes a surprise.

````python
# Stacking files that do not have identical columns
partial = pd.DataFrame({"order_id": [2001], "item": ["Cable"], "new_column": ["extra"]})
combined = pd.concat([orders.head(2), partial], ignore_index=True)

print(combined[["order_id", "item", "unit_price", "new_column"]])
print()
print("missing columns become NaN on both sides. concat does not complain.")
````

That silence is the danger. If a source file changes its column names, `concat`
happily produces a table where half the rows have nulls in the old column and half
in the new. Check your columns match before stacking.

````python
# A quick guard worth stealing
def safe_concat(frames, names=None):
    """Stack DataFrames, refusing if their columns do not match."""
    names = names or [f"frame_{i}" for i in range(len(frames))]
    base = set(frames[0].columns)
    for name, frame in zip(names[1:], frames[1:]):
        if set(frame.columns) != base:
            missing = base - set(frame.columns)
            extra = set(frame.columns) - base
            raise ValueError(f"{name}: missing {missing or 'none'}, extra {extra or 'none'}")
    return pd.concat(frames, ignore_index=True)


print(len(safe_concat([orders, may], ["april", "may"])))

try:
    safe_concat([orders, partial], ["orders", "partial"])
except ValueError as e:
    print(f"refused: {str(e)[:100]}")
````

## Practice: Transforming columns

### Vectorised first, `apply` second, loops never

Pandas operates on whole columns at once. That is the point of it. Writing a Python
loop over rows gives up the only advantage pandas has.

````python
import pandas as pd, time
orders = pd.read_csv("orders.csv", parse_dates=["order_date"])

# Vectorised: one operation on the whole column
orders["revenue"] = orders["quantity"] * orders["unit_price"]
orders["vat"] = orders["revenue"] * 0.075
orders["total"] = orders["revenue"] + orders["vat"]

orders[["order_id", "quantity", "unit_price", "revenue", "vat", "total"]].head()
````

````python
# How much does it matter? Build a bigger frame and time all three approaches.
big = pd.concat([orders] * 4000, ignore_index=True)
print(f"{len(big):,} rows")

start = time.perf_counter()
result_vec = big["quantity"] * big["unit_price"]
t_vec = time.perf_counter() - start

start = time.perf_counter()
result_apply = big.apply(lambda row: row["quantity"] * row["unit_price"], axis=1)
t_apply = time.perf_counter() - start

start = time.perf_counter()
values = []
for _, row in big.iterrows():
    values.append(row["quantity"] * row["unit_price"])
t_loop = time.perf_counter() - start

print(f"vectorised:      {t_vec * 1000:8.1f} ms   (1x)")
print(f"apply(axis=1):   {t_apply * 1000:8.1f} ms   ({t_apply / t_vec:.0f}x slower)")
print(f"iterrows loop:   {t_loop * 1000:8.1f} ms   ({t_loop / t_vec:.0f}x slower)")
````

Same answer, wildly different cost. On a few hundred rows nobody notices. On ten
million, vectorised finishes while `iterrows` is still going at lunchtime.

**Reach for `apply` only when the logic genuinely cannot be expressed as column
arithmetic.** Reach for `iterrows` essentially never.

````python
# apply on a single column is fine and readable
orders["price_band"] = orders["unit_price"].apply(
    lambda price: "unknown" if pd.isna(price) else ("high" if price > 100000 else "low")
)
print(orders[["order_id", "unit_price", "price_band"]].head(6))
````

````python
# But np.where or np.select is faster and clearer for conditional columns
import numpy as np

orders["band"] = np.where(orders["unit_price"] > 100000, "high", "low")

conditions = [
    orders["unit_price"].isna(),
    orders["unit_price"] > 400000,
    orders["unit_price"] > 50000,
]
labels = ["unknown", "premium", "mid"]
orders["tier"] = np.select(conditions, labels, default="budget")

orders[["order_id", "unit_price", "band", "tier"]].head(8)
````

`np.select` reads top to bottom like an `if/elif/else` chain, and the first matching
condition wins. It is the pandas equivalent of a SQL `CASE WHEN`.

````python
# map() on a Series: swap values using a dictionary
country_names = {"NG": "Nigeria", "GH": "Ghana", "KE": "Kenya"}
orders["country_full"] = orders["country"].map(country_names)

print(orders[["country", "country_full"]].drop_duplicates())
````

Anything not in the dictionary becomes `NaN`. That is useful: it tells you about
codes you did not know existed. Use `.map(d).fillna("unknown")` if you would rather
not lose them.

````python
# DataFrame.map applies to every cell.
# Note: the old name was applymap. It was deprecated in pandas 2.1 and REMOVED in 3.0.
numeric = orders[["quantity", "unit_price"]]

try:
    doubled = numeric.map(lambda x: x * 2)              # pandas 2.1+
    print("used DataFrame.map")
except AttributeError:
    doubled = numeric.applymap(lambda x: x * 2)         # older pandas
    print("fell back to applymap (old pandas)")

doubled.head(3)
````

If you have `df.applymap(...)` anywhere in your own code, change it to `df.map(...)`
now. It still works with a warning on pandas 2.x and fails outright on 3.x.

````python
# Renaming, dropping, reordering
renamed = orders.rename(columns={"unit_price": "price", "order_date": "date"})
print(list(renamed.columns)[:6])

dropped = orders.drop(columns=["band", "tier", "price_band"])
print(f"{orders.shape[1]} columns -> {dropped.shape[1]} columns")

reordered = orders[["order_id", "order_date", "customer_id", "item", "revenue"]]
reordered.head(3)
````

````python
# assign(): build several columns in one readable chain
report = (
    pd.read_csv("orders.csv", parse_dates=["order_date"])
    .assign(
        revenue=lambda d: d["quantity"] * d["unit_price"],
        vat=lambda d: d["revenue"] * 0.075,
        month=lambda d: d["order_date"].dt.strftime("%Y-%m"),
        is_large=lambda d: d["revenue"] > 200000,
    )
    .query("status == 'completed'")
    .sort_values("revenue", ascending=False)
    .head(5)
)

report[["order_id", "item", "revenue", "vat", "month", "is_large"]]
````

Each `lambda d:` receives the DataFrame as it exists at that point in the chain,
so `vat` can use the `revenue` that the line above just created. Chains like this
read top to bottom as a list of steps, and there are no intermediate variables to

The default writes the row numbers as an unnamed first column. Read that file back
and you get a mystery column called `Unnamed: 0`. Pass `index=False` unless the index
genuinely carries meaning, such as a date index on a time series.

````python
# The formats, and when each is right
orders.to_csv("out.csv", index=False)
orders.to_json("out.json", orient="records", indent=2, date_format="iso")
orders.to_json("out.jsonl", orient="records", lines=True, date_format="iso")

try:
    orders.to_parquet("out.parquet", index=False)
    parquet_ok = True
except Exception as e:
    parquet_ok = False
    print(f"parquet needs pyarrow: {type(e).__name__}")

import os
for name in ["out.csv", "out.json", "out.jsonl"] + (["out.parquet"] if parquet_ok else []):
    print(f"{name:<14} {os.path.getsize(name):>7} bytes")
````

| Format | Use it for | Watch out |
|---|---|---|
| CSV | Sharing with humans, Excel, anything | No types. Everything comes back as text |
| JSON | Config, API payloads, nested data | Whole file must be read at once |
| JSONL | Events, logs, append-only streams | One object per line |
| **Parquet** | **Anything going into a warehouse** | Needs `pyarrow`, not human readable |

Parquet keeps your dtypes, compresses far smaller, and lets a query engine read just
the columns it needs. Once data leaves the exploration stage, Parquet is the default
in every warehouse you will work with. CSV is for handing to a person.

````python
# Writing partitioned output, the shape a warehouse expects
import os

df = orders.dropna(subset=["order_date"]).copy()
df["partition"] = df["order_date"].dt.strftime("%Y-%m")

for partition, group in df.groupby("partition"):
    folder = os.path.join("output", f"month={partition}")
    os.makedirs(folder, exist_ok=True)
    path = os.path.join(folder, "orders.csv")
    group.drop(columns=["partition"]).to_csv(path, index=False)
    print(f"{path}  ({len(group)} rows)")
````

`month=2026-01` as a folder name is Hive-style partitioning. Every engine you will
meet, from Spark to BigQuery to DuckDB, reads that layout and can skip whole folders
when your query filters on the month. Writing it this way is free, and it makes
queries that scan a year of data read only the months they need.

````python
# Reading a whole partitioned folder back in
import glob

frames = []
for path in sorted(glob.glob("output/**/*.csv", recursive=True)):
    part = pd.read_csv(path)
    part["source_file"] = path
    frames.append(part)

combined = pd.concat(frames, ignore_index=True)
print(f"{len(frames)} file(s) -> {len(combined)} rows")
print(combined["source_file"].value_counts())
````

## Pandas to SQL, side by side

You already know SQL. This table is the shortest path to being useful in pandas.

| SQL | pandas |
|---|---|
| `SELECT * FROM t` | `df` |
| `SELECT a, b FROM t` | `df[["a", "b"]]` |
| `SELECT * FROM t LIMIT 5` | `df.head(5)` |
| `SELECT DISTINCT a FROM t` | `df["a"].unique()` or `df.drop_duplicates(subset=["a"])` |
| `COUNT(DISTINCT a)` | `df["a"].nunique()` |
| `WHERE a > 10` | `df[df["a"] > 10]` |
| `WHERE a > 10 AND b = 'x'` | `df[(df["a"] > 10) & (df["b"] == "x")]` |
| `WHERE a IN (1, 2)` | `df[df["a"].isin([1, 2])]` |
| `WHERE a BETWEEN 1 AND 9` | `df[df["a"].between(1, 9)]` |
| `WHERE a IS NULL` | `df[df["a"].isna()]` |
| `WHERE a LIKE '%lap%'` | `df[df["a"].str.contains("lap", na=False)]` |
| `ORDER BY a DESC` | `df.sort_values("a", ascending=False)` |
| `GROUP BY a` | `df.groupby("a")` |
| `SELECT a, SUM(b) FROM t GROUP BY a` | `df.groupby("a")["b"].sum().reset_index()` |
| `HAVING SUM(b) > 100` | `.groupby("a").filter(lambda g: g["b"].sum() > 100)` |
| `SUM(b) OVER (PARTITION BY a)` | `df.groupby("a")["b"].transform("sum")` |
| `JOIN ... ON id` | `left.merge(right, on="id", how="inner")` |
| `LEFT JOIN` | `how="left"` |
| `UNION ALL` | `pd.concat([df1, df2], ignore_index=True)` |
| `CASE WHEN ... THEN ...` | `np.select(conditions, labels, default=...)` |
| `COALESCE(a, 0)` | `df["a"].fillna(0)` |
| `CAST(a AS FLOAT)` | `pd.to_numeric(df["a"], errors="coerce")` |
| `CREATE TABLE AS ...` | `df.to_parquet(path)` |

Two habits SQL does not teach you, that pandas needs:

- **`.copy()`** when you filter and then intend to modify the result
- **`validate=`** on every merge, because pandas will happily fan out your rows

## Common mistakes to watch for

| Mistake | What you see | The fix |
|---|---|---|
| `and` / `or` in a filter | "truth value of a Series is ambiguous" | `&`, `\|`, `~` |
| Missing parentheses in a filter | Confusing `TypeError` | `(df["a"] > 1) & (df["b"] == 2)` |
| `df[mask]["col"] = x` | `SettingWithCopyWarning`, change may not stick | `df.loc[mask, "col"] = x` |
| Filtering then assigning | Same warning | `df[mask].copy()` first |
| `df.shape()` | `TypeError: not callable` | `df.shape`, no parentheses |
| `df.head` | Prints `<bound method ...>` | `df.head()`, with parentheses |
| `loc[0:3]` expecting 3 rows | You get 4 | `loc` includes the stop, `iloc` does not |
| `df.mean()` on mixed types | `TypeError` | `df.mean(numeric_only=True)` |
| Bare `dropna()` | Most of your data disappears | `dropna(subset=["the_column"])` |
| `astype(float)` on messy text | `ValueError` on one bad row | `pd.to_numeric(..., errors="coerce")` |
| Merge without `validate=` | Row count quietly multiplies | `validate="many_to_one"` |
| `inner` join by default habit | Fact rows silently vanish | `how="left"` unless you mean otherwise |
| `concat` with mismatched columns | Half your rows full of nulls | Check columns match first |
| `concat` without `ignore_index` | Duplicate index values | `ignore_index=True` |
| `to_csv()` without `index=False` | An `Unnamed: 0` column next time | `index=False` |
| `iterrows` for arithmetic | 100x slower than it needs to be | Vectorise, or `apply` at worst |
| `df.applymap` | Deprecated in 2.1, gone in 3.0 | `df.map` |
| Chained indexing `df["a"]["b"]` | Unpredictable | `df.loc[row, "col"]` |

````python
# One last habit: check your row count after every merge. Every single time.
import pandas as pd

orders = pd.read_csv("orders.csv")
customers = pd.read_csv("customers.csv")

before = len(orders)
merged = orders.merge(customers, on="customer_id", how="left", validate="many_to_one")
after = len(merged)

print(f"before: {before}, after: {after}")
if after != before:
    print("STOP. A left join must never change the row count. Your right table has duplicate keys.")
else:
    print("row count preserved, join is safe")
````
