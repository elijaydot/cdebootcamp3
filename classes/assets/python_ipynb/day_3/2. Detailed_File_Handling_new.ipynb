````python
import os, json, random

os.makedirs("datasets", exist_ok=True)

# 1. the plain text file the early examples read
with open("example.txt", "w", encoding="utf-8") as f:
    f.write("Lorem ipsum dolor sit amet\n")
    f.write("ahmed is a chelsea fan\n")
    f.write("This will append to the file.")

# 2. a small clean CSV
with open("sales.csv", "w", encoding="utf-8") as f:
    f.write("order_id,customer,item,amount,country\n")
    f.write("1,ahmed,laptop,450000,NG\n")
    f.write("2,lahla,phone,180000,GH\n")
    f.write("3,ahmed,mouse,8000,NG\n")
    f.write("4,judith,laptop,520000,NG\n")
    f.write("5,uche,monitor,95000,KE\n")

# 3. a messy CSV, the kind you actually receive
with open("messy_sales.csv", "w", encoding="utf-8") as f:
    f.write("order_id,customer,item,amount,country\n")
    f.write("6,ahmed,keyboard,15000,NG\n")
    f.write("7,,phone,180000,GH\n")          # missing customer
    f.write("8,judith,laptop,N/A,NG\n")      # unparseable amount
    f.write("9,uche,monitor,,KE\n")          # empty amount
    f.write("10,lahla,mouse,7500,\n")        # missing country
    f.write("6,ahmed,keyboard,15000,NG\n")   # duplicate of row 6

# 4. a JSON config
config = {
    "source": "postgres",
    "tables": ["customers", "orders", "payments"],
    "batch_size": 5000,
    "checks": {"null_check": True, "row_count_check": True},
}
with open("config.json", "w", encoding="utf-8") as f:
    json.dump(config, f, indent=2)

# 5. a bigger sales file in datasets/, so the pandas cell further down works
random.seed(42)
regions = ["Sub-Saharan Africa", "Europe", "Middle East and North Africa", "Asia"]
countries = ["Chad", "Latvia", "Pakistan", "Czech Republic", "Nigeria", "Kenya", "Ghana"]
items = ["Office Supplies", "Beverages", "Vegetables", "Household", "Cereal"]

with open("datasets/another_sales_records.csv", "w", encoding="utf-8") as f:
    f.write("Region,Country,Item Type,Sales Channel,Order Priority,Order Date,Order ID,"
            "Ship Date,Units Sold,Unit Price,Unit Cost,Total Revenue,Total Cost,Total Profit\n")
    for i in range(60):
        units = random.randint(500, 9000)
        price = round(random.uniform(30, 700), 2)
        cost = round(price * random.uniform(0.6, 0.85), 2)
        revenue = round(units * price, 2)
        total_cost = round(units * cost, 2)
        f.write(
            f"{random.choice(regions)},{random.choice(countries)},{random.choice(items)},"
            f"{random.choice(['Online', 'Offline'])},{random.choice(['L', 'M', 'H', 'C'])},"
            f"{random.randint(1, 12)}/{random.randint(1, 28)}/{random.randint(2011, 2016)},"
            f"{random.randint(100000000, 999999999)},"
            f"{random.randint(1, 12)}/{random.randint(1, 28)}/{random.randint(2011, 2016)},"
            f"{units},{price},{cost},{revenue},{total_cost},{round(revenue - total_cost, 2)}\n"
        )

print("Created:")
for name in ["example.txt", "sales.csv", "messy_sales.csv", "config.json",
             "datasets/another_sales_records.csv"]:
    print(f"  {name}  ({os.path.getsize(name)} bytes)")
````

## Using `with` statement
The `with` statement is the preferred way to work with files in Python because it ensures that the file is properly closed after the block of code is executed, even if an error occurs.

````python
# Using 'with' to open a file
with open('example.txt', 'r') as file:
    contents = file.read()
    print(contents)
# File is automatically closed outside the 'with' block
````

### Why `with` is better?
- **Automatic file closure**: You don't have to manually close the file, reducing the chance of errors.
- **Error handling**: Even if an exception occurs within the `with` block, the file will still be closed properly.


## Writing to Files
You can write data to files using the `w` (write) or `a` (append) modes.
- **Write mode (`w`)**: Overwrites the file if it already exists.
- **Append mode (`a`)**: Appends data to the file if it already exists.

````python
# Writing to a file (overwrite mode 'w')
with open('jerry.txt', 'a') as file:
    file.write('\n we are still waiting ooo...oga you never stil share money.')
````

## Reading Files
There are several ways to read files, depending on your needs:
- `read()`: Reads the entire file.
- `readline()`: Reads one line at a time.
- `readlines()`: Reads all lines and returns them as a list.

````python
# Reading files using read(), readline(), and readlines()
with open('example.txt', 'r') as file:
    # Read the entire file
    contents = file.read()
    print("Read entire file:", contents)
    
    # Reset the file pointer to the beginning
    file.seek(0)
    
    # Read one line
    first_line = file.readline()
    print("First line:", first_line)
    
    # Read all lines as a list
    file.seek(0)
    all_lines = file.readlines()
    print("All lines:", all_lines)
````

## Working with Binary Files
Binary files contain data in binary format (0s and 1s), such as images or executable files. You need to open them in binary mode (`rb`, `wb`, etc.).

````python
# Example of reading a binary file (image)
with open('example_image.png', 'rb') as binary_file:
    binary_data = binary_file.read()
    print("Binary data:", binary_data[:10])  # Print first 10 bytes
````

## Closing Files
It's crucial to close files after opening them to free up system resources. This is done automatically when using the `with` statement, but if you're not using `with`, always remember to call `file.close()`.

````python
# Opening a file without 'with' and manually closing it
file = open('example.txt', 'r')
contents = file.read()
print(contents)
file.close()  # Always remember to close the file
````

## File Iteration
You can iterate over the lines in a file using a loop. This is especially useful for large files that can’t be loaded entirely into memory.

## Practice: Reading and Writing

### Always use `with`. Here is what happens when you do not.

````python
# The manual way. Notice what is missing.
file = open("example.txt", "r", encoding="utf-8")
contents = file.read()
print(contents)
print(f"closed? {file.closed}")      # False. The file is still open.
file.close()
print(f"closed now? {file.closed}")
````

Earlier in this notebook there is a cell that opens `example.txt` and calls
`readlines()` but never closes it. That file handle stays open until the kernel
restarts. One file is harmless. A loop over 5,000 files will hit
`OSError: Too many open files` and take your job down.

`with` closes the file for you, even if an error is raised inside the block.

````python
with open("example.txt", "r", encoding="utf-8") as file:
    contents = file.read()

print(f"closed automatically? {file.closed}")
print(contents)
````

````python
# It closes even when the code inside blows up
try:
    with open("example.txt", "r", encoding="utf-8") as file:
        contents = file.read()
        raise ValueError("something went wrong mid-processing")
except ValueError as e:
    print(f"caught: {e}")

print(f"file still closed properly? {file.closed}")
````

### The four ways to read, and when each is right

| Method | Gives you | Memory | Use when |
|---|---|---|---|
| `.read()` | One big string | Whole file in RAM | Small files, or you need the whole text |
| `.readline()` | One line | One line | You want just the header |
| `.readlines()` | A list of lines | Whole file in RAM | Small files, you need a list |
| `for line in file` | One line at a time | One line | **Large files. This is the default in DE** |

````python
with open("example.txt", "r", encoding="utf-8") as f:
    print("read():")
    print(repr(f.read()))

with open("example.txt", "r", encoding="utf-8") as f:
    print()
    print("readline() twice:")
    print(repr(f.readline()))
    print(repr(f.readline()))

with open("example.txt", "r", encoding="utf-8") as f:
    print()
    print("readlines():")
    print(f.readlines())
````

Notice the `\n` at the end of every line except the last. `repr()` shows them,
`print()` hides them. Those hidden newlines are behind most "why is my data wrong"
moments in file handling, which is why `.strip()` exists.

````python
# Iterating: the only one that works on a 50GB file
with open("example.txt", "r", encoding="utf-8") as f:
    for line_number, line in enumerate(f, start=1):
        print(f"{line_number}: {line.strip()}")
````

Why it matters: `.read()` on a 50GB file tries to put 50GB in memory and your
machine dies. Iterating holds one line at a time, so the file size stops mattering.
Default to iteration and only reach for `.read()` when you know the file is small.

````python
# seek(0) rewinds. A file can only be read once without it.

with open("example.txt", "r", encoding="utf-8") as f:
    first = f.read()
    second = f.read()          # empty. the cursor is already at the end
    print(f"first read:  {len(first)} characters")
    print(f"second read: {len(second)} characters")

    f.seek(0)                  # back to the start
    third = f.read()
    print(f"after seek:  {len(third)} characters")
````

### Writing: `w` destroys, `a` adds

````python
# w creates the file, or wipes it completely if it already exists
with open("notes.txt", "w", encoding="utf-8") as f:
    f.write("first line\n")
    f.write("second line\n")

with open("notes.txt", "r", encoding="utf-8") as f:
    print(f.read())
````

````python
# Run this cell twice and watch what happens
with open("notes.txt", "w", encoding="utf-8") as f:
    f.write("this replaced everything\n")

with open("notes.txt", "r", encoding="utf-8") as f:
    print(f.read())
````

````python
# a appends instead
with open("notes.txt", "a", encoding="utf-8") as f:
    f.write("this was added\n")
    f.write("and this\n")

with open("notes.txt", "r", encoding="utf-8") as f:
    print(f.read())
````

**`w` is destructive and silent about it.** There is no "are you sure". Opening a
file in `w` mode by mistake is one of the few genuinely unrecoverable things you can
do in a notebook. When you are not certain, use `x` mode: it creates the file and
raises `FileExistsError` if something is already there.

````python
# x mode refuses to overwrite
try:
    with open("notes.txt", "x", encoding="utf-8") as f:
        f.write("this will never be written")
except FileExistsError as e:
    print(f"FileExistsError: {e}")
    print("good. your existing file is safe")
````

````python
# write() does NOT add newlines. writelines() does not either. This surprises everyone.

with open("no_newlines.txt", "w", encoding="utf-8") as f:
    f.write("one")
    f.write("two")
    f.write("three")

with open("no_newlines.txt", "r", encoding="utf-8") as f:
    print(repr(f.read()))          # 'onetwothree'

# Two ways to do it properly
lines = ["one", "two", "three"]

with open("with_newlines.txt", "w", encoding="utf-8") as f:
    for line in lines:
        f.write(line + "\n")

with open("with_newlines2.txt", "w", encoding="utf-8") as f:
    f.write("\n".join(lines))     # no trailing newline, often what you want

with open("with_newlines.txt", "r", encoding="utf-8") as f:
    print(repr(f.read()))
````

### Encoding: the bug that only shows up in production

Always pass `encoding="utf-8"`. Without it, Python uses whatever your operating
system defaults to, which differs between your laptop and the server. That is how a
file with an accent or a naira sign works locally and crashes in the pipeline.

````python
# Names and symbols that are not plain ASCII
with open("names.txt", "w", encoding="utf-8") as f:
    f.write("Adébáyò\n")
    f.write("Noisy-le-Grand, Île-de-France\n")
    f.write("Amount: ₦450,000\n")

with open("names.txt", "r", encoding="utf-8") as f:
    print(f.read())

print(f"file size on disk: {os.path.getsize('names.txt')} bytes")
print("(more bytes than characters, because accented letters take 2 bytes in utf-8)")
````

````python
# When a file has a stray byte you cannot control, do not let it kill the job
with open("broken.txt", "wb") as f:                 # wb = write bytes
    f.write(b"good line\n\xff\xfe broken bytes\ngood again\n")

try:
    with open("broken.txt", "r", encoding="utf-8") as f:
        print(f.read())
except UnicodeDecodeError as e:
    print(f"UnicodeDecodeError: {e}")

print()
print("with errors='replace', the job survives:")
with open("broken.txt", "r", encoding="utf-8", errors="replace") as f:
    print(f.read())
````

`errors="replace"` swaps the bad bytes for a placeholder instead of raising.
Use it when finishing the run matters more than perfect fidelity, and log that you
did. `errors="ignore"` drops them silently, which is usually worse.

### CSV the manual way, and why you should not

Splitting on commas looks fine until a field contains a comma.

````python
# The naive approach
with open("sales.csv", "r", encoding="utf-8") as f:
    for line in f:
        parts = line.strip().split(",")
        print(parts)
````

````python
# Now watch it break on a perfectly normal file
with open("tricky.csv", "w", encoding="utf-8") as f:
    f.write("id,company,amount\n")
    f.write('1,"Dangote Cement, Plc",5000\n')     # a comma INSIDE a quoted field
    f.write("2,Quidax,1200\n")

print("naive split:")
with open("tricky.csv", "r", encoding="utf-8") as f:
    next(f)                                        # skip the header
    for line in f:
        print(line.strip().split(","))             # 4 fields instead of 3

print()
print("csv module:")
import csv
with open("tricky.csv", "r", encoding="utf-8", newline="") as f:
    reader = csv.reader(f)
    next(reader)
    for row in reader:
        print(row)                                 # 3 fields, correctly
````

The `csv` module understands quoting, embedded commas, embedded newlines and
escaped quotes. Never write your own CSV parser. This is not a style preference,
it is the difference between a pipeline that works and one that corrupts silently.

````python
# csv.DictReader: every row becomes a dictionary, keyed by the header. This is the one you want.
import csv

with open("sales.csv", "r", encoding="utf-8", newline="") as f:
    reader = csv.DictReader(f)
    rows = list(reader)

print(f"{len(rows)} rows")
print(rows[0])
print()
for row in rows:
    print(f"{row['order_id']}: {row['customer']} bought a {row['item']} for {row['amount']}")
````

Look at that output carefully. `row["amount"]` is the **string** `"450000"`, not the
number. CSV has no types. Everything arrives as text and converting it is your job.
This is the number one source of wrong totals in beginner pipelines.

````python
# Proof, and the fix
total_wrong = ""
total_right = 0

for row in rows:
    total_wrong += row["amount"]          # string concatenation
    total_right += int(row["amount"])     # actual addition

print(f"wrong: {total_wrong[:40]}...")
print(f"right: {total_right:,}")
````

````python
# csv.DictWriter: writing a list of dicts back out
import csv

clean_rows = [
    {"order_id": 1, "customer": "ahmed", "amount": 450000},
    {"order_id": 2, "customer": "lahla", "amount": 180000},
]

with open("output.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=["order_id", "customer", "amount"])
    writer.writeheader()
    writer.writerows(clean_rows)

with open("output.csv", "r", encoding="utf-8") as f:
    print(f.read())
````

`newline=""` in the `open()` call is not optional. Leave it out and on Windows every
row gets a blank line between it. The `csv` module handles line endings itself.

````python
# JSON: one object per file
import json

with open("config.json", "r", encoding="utf-8") as f:
    config = json.load(f)              # load reads from a FILE

print(type(config).__name__)
print(config["source"])
print(config["tables"])
print(config["checks"]["null_check"])

# loads (with an s) reads from a STRING
text = '{"id": 1, "user": "ahmed"}'
record = json.loads(text)
print(record["user"])
````

Remember the pair: `load`/`dump` work on **files**, `loads`/`dumps` work on
**strings**. The `s` stands for string. Everyone mixes these up for the first month.

````python
# Writing JSON back out
import json

summary = {
    "run_date": "2026-09-14",
    "rows_in": 120,
    "rows_loaded": 115,
    "rejected": [{"id": 7, "reason": "empty customer"}],
}

with open("run_summary.json", "w", encoding="utf-8") as f:
    json.dump(summary, f, indent=2)

with open("run_summary.json", "r", encoding="utf-8") as f:
    print(f.read())
````

````python
# JSONL: one JSON object per LINE. The format of choice for event data.
import json

events = [
    {"event": "login",  "user": "ahmed",  "ts": "2026-09-14T08:00:00"},
    {"event": "trade",  "user": "ahmed",  "ts": "2026-09-14T08:05:00", "amount": 5000},
    {"event": "logout", "user": "lahla",  "ts": "2026-09-14T08:11:00"},
]

with open("events.jsonl", "w", encoding="utf-8") as f:
    for event in events:
        f.write(json.dumps(event) + "\n")

print("raw file:")
with open("events.jsonl", "r", encoding="utf-8") as f:
    print(f.read())

print("reading it back, one line at a time:")
with open("events.jsonl", "r", encoding="utf-8") as f:
    for line in f:
        event = json.loads(line)
        print(f"  {event['event']} by {event['user']}")
````

Why JSONL and not one big JSON array? Because you can read it line by line without
loading the whole thing, you can append a new event without rewriting the file, and
one corrupt line does not destroy the rest. Every logging and streaming system you
will meet uses it.

## Practice: File errors, and writing files safely

### The errors you will actually meet

| Exception | When |
|---|---|
| `FileNotFoundError` | Reading a file that is not there. The upstream job has not landed yet |
| `FileExistsError` | `x` mode on a file that already exists |
| `PermissionError` | No rights to read or write, or the path is a folder |
| `IsADirectoryError` | You passed a folder where a file was expected |
| `UnicodeDecodeError` | Wrong encoding, or the file is actually binary |
| `ValueError` | An invalid mode string, such as `open(path, "")` |

There is a cell earlier in this notebook with `open('non_existent_file.txt', '')`.
The empty mode raises `ValueError`, not `FileNotFoundError`. Worth knowing, because
the error you get tells you which mistake you made.

````python
# Catch what you can actually handle, and say something useful
def read_file_safely(path):
    """Return the file contents, or None with a clear message."""
    try:
        with open(path, "r", encoding="utf-8") as f:
            return f.read()
    except FileNotFoundError:
        print(f"  {path}: not found. has the upstream job run?")
    except IsADirectoryError:
        print(f"  {path}: that is a folder, not a file")
    except PermissionError:
        print(f"  {path}: no permission to read")
    except UnicodeDecodeError:
        print(f"  {path}: not valid utf-8. is it a binary file?")
    return None


for path in ["example.txt", "missing.txt", "datasets", "broken.txt"]:
    print(path)
    result = read_file_safely(path)
    if result is not None:
        print(f"  read {len(result)} characters")
````

Compare that with the bare `except:` earlier in this notebook, which prints
"The file does not exist" no matter what actually went wrong. If the real problem is
a permissions issue, you will spend an hour looking for a missing file that is
sitting right there.

````python
# Check first, or ask forgiveness? Both are fine. Know why.
import os

path = "sales.csv"

# Look before you leap
if os.path.exists(path):
    with open(path, encoding="utf-8") as f:
        print(f"LBYL: read {len(f.read())} characters")
else:
    print("LBYL: file is not there")

# Easier to ask forgiveness than permission. The Python way, and safer.
try:
    with open(path, encoding="utf-8") as f:
        print(f"EAFP: read {len(f.read())} characters")
except FileNotFoundError:
    print("EAFP: file is not there")
````

Prefer the `try` version. Between your `os.path.exists()` check and your `open()`,
another process can delete the file. It is rare, and it will happen to you exactly
once, at the worst possible time.

````python
# The atomic write pattern: never leave a half-written file behind
import os, csv

def write_csv_atomically(path, rows, fieldnames):
    """Write to a temp file first, then rename. Readers never see a partial file."""
    temp_path = path + ".tmp"
    with open(temp_path, "w", encoding="utf-8", newline="") as f:
        writer = csv.DictWriter(f, fieldnames=fieldnames)
        writer.writeheader()
        writer.writerows(rows)
    os.replace(temp_path, path)          # rename is atomic on the same filesystem
    return path


rows = [{"id": 1, "amount": 500}, {"id": 2, "amount": 900}]
write_csv_atomically("safe_output.csv", rows, ["id", "amount"])

with open("safe_output.csv", encoding="utf-8") as f:
    print(f.read())
````

Why bother? If your job crashes halfway through writing `sales.csv`, the naive
version leaves a truncated file on disk that looks complete. Whatever reads it next
gets half the data and no error. Writing to `.tmp` and renaming at the end means the
real filename either has the old complete file or the new complete file, never a
half one. Every serious pipeline does this.

````python
# Processing a large file safely, one line at a time, with per-row error handling
import csv

loaded = 0
skipped = 0
total = 0

with open("messy_sales.csv", "r", encoding="utf-8", newline="") as f:
    reader = csv.DictReader(f)
    for line_number, row in enumerate(reader, start=2):    # start=2, line 1 is the header
        try:
            amount = float(row["amount"])
            if not row["customer"]:
                raise ValueError("empty customer")
            total += amount
            loaded += 1
        except ValueError as e:
            skipped += 1
            print(f"  line {line_number}: skipped, {e}")

print()
print(f"loaded {loaded}, skipped {skipped}, total {total:,.2f}")
````

`enumerate(reader, start=2)` gives you the real line number in the file, because
line 1 was the header. When you report a bad row, the person fixing it needs the
line number, not the row index.

## Common mistakes to watch for

| Mistake | What you see | The fix |
|---|---|---|
| `open()` without `with` | File handles leak, eventually `OSError` | Always use `with` |
| Opening in `w` when you meant `a` | The file is silently wiped | `a` to append, `x` to refuse to overwrite |
| No `encoding="utf-8"` | Works locally, `UnicodeDecodeError` on the server | Always pass it |
| `.read()` on a huge file | The kernel dies, or swaps for ten minutes | `for line in file:` |
| Reading twice without `seek(0)` | The second read is empty | `f.seek(0)` to rewind |
| `line.split(",")` for CSV | Breaks on any quoted comma | `csv.reader` or `csv.DictReader` |
| No `newline=""` with csv | Blank line between every row on Windows | Pass `newline=""` |
| Forgetting values from CSV are strings | `"1"+"2"` gives `"12"`, totals are nonsense | Convert with `int()` / `float()` |
| Forgetting `\n` in `write()` | Everything on one line | `f.write(line + "\n")` |
| Confusing `load` and `loads` | `TypeError` about str vs file | `load` = file, `loads` = string |
| `os.makedirs` without `exist_ok` | Crashes on the second run | `os.makedirs(path, exist_ok=True)` |
| Bare `except:` around file work | "File not found" when it was permissions | Name the exception you expect |
| Writing straight to the final path | A crash leaves a half file that looks whole | Write `.tmp`, then `os.replace` |
