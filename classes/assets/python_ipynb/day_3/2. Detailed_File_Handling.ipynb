# File Handling

File handling is an essential concept in Python for reading, writing, and manipulating files. Python provides several ways to work with files, both text and binary.

## Opening Files
Files can be opened using the `open()` function. You need to specify the file path and the mode in which to open the file.

### Common Modes:
- `r`: Read (default mode, opens file for reading).
- `w`: Write (opens file for writing, truncates if exists).
- `a`: Append (opens file for appending).
- `b`: Binary mode (can be added to any of the above modes).
- `x`: Create (creates file, fails if file exists).


```python
# Example of opening a file for reading (default mode 'r')
file = open('example.txt', 'r')
contents = file.read()
print(contents)
file.close()
```

```python
with open('example.txt', 'r') as file:
    contents = file.read()
    print(contents)
# File is automatically closed outside the 'with' block
```

```python
file = open('example.txt', 'r')
lines = file.readlines()
for line in lines:
    print(line)
```

## Using `with` statement
The `with` statement is the preferred way to work with files in Python because it ensures that the file is properly closed after the block of code is executed, even if an error occurs.

```python
# Using 'with' to open a file
with open('example.txt', 'r') as file:
    contents = file.read()
    print(contents)
# File is automatically closed outside the 'with' block
```

### Why `with` is better?
- **Automatic file closure**: You don't have to manually close the file, reducing the chance of errors.
- **Error handling**: Even if an exception occurs within the `with` block, the file will still be closed properly.


## Writing to Files
You can write data to files using the `w` (write) or `a` (append) modes.
- **Write mode (`w`)**: Overwrites the file if it already exists.
- **Append mode (`a`)**: Appends data to the file if it already exists.

```python
# Writing to a file (overwrite mode 'w')
with open('salome.txt', 'a') as file:
    file.write('\n we are still waiting ooo...chief Osas.')
```

## Reading Files
There are several ways to read files, depending on your needs:
- `read()`: Reads the entire file.
- `readline()`: Reads one line at a time.
- `readlines()`: Reads all lines and returns them as a list.

```python
# Reading files using read(), readline(), and readlines()
with open('example.txt', 'r') as file:
    # Read the entire file
    contents = file.read()
    print("Read entire file:", contents)
    
    # Reset the file pointer to the beginning
    file.seek(0)
    
    # Read one line
    first_line = file.readline()
    print("First line:", first_line)
    
    # Read all lines as a list
    file.seek(0)
    all_lines = file.readlines()
    print("All lines:", all_lines)
```

## Handling File Exceptions
When working with files, it’s important to handle potential exceptions, such as the file not existing or trying to write to a read-only file.

```python
with open('non_existent_file.txt', 'r') as file:
        contents = file.read()
```

```python
try:
    with open('non_existent_file.txt', 'r') as file:
        contents = file.read()
except FileNotFoundError:
    print("The file does not exist.")
except IOError:
    print("An IOError occurred.")
```

```python
try:
    with open('non_existent_file.txt', 'r') as file:
        contents = file.read()
except:
    print("The file does not exist.")
```

## Working with Binary Files
Binary files contain data in binary format (0s and 1s), such as images or executable files. You need to open them in binary mode (`rb`, `wb`, etc.).

```python
# Example of reading a binary file (image)
with open('example_image.png', 'rb') as binary_file:
    binary_data = binary_file.read()
    print("Binary data:", binary_data[:10])  # Print first 10 bytes
```

## Closing Files
It's crucial to close files after opening them to free up system resources. This is done automatically when using the `with` statement, but if you're not using `with`, always remember to call `file.close()`.

```python
# Opening a file without 'with' and manually closing it
file = open('example.txt', 'r')
contents = file.read()
print(contents)
file.close()  # Always remember to close the file
```

## File Iteration
You can iterate over the lines in a file using a loop. This is especially useful for large files that can’t be loaded entirely into memory.

```python
# Iterating over the lines of a file
with open('example.txt', 'r') as file:
    for line in file:
        print(line.strip())  # strip() removes leading/trailing whitespace
```
