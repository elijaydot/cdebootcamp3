## Lists: DE tasks

```python
transactions = [
    {"id": 1, "user": "ahmed",  "amount": 5000,  "currency": "NGN", "status": "success"},
    {"id": 2, "user": "lahla",  "amount": 120,   "currency": "USD", "status": "failed"},
    {"id": 3, "user": "judith", "amount": 900,   "currency": "NGN", "status": "success"},
    {"id": 4, "user": "uche",   "amount": 4300,  "currency": "NGN", "status": "success"},
    {"id": 5, "user": "ahmed",  "amount": 80,    "currency": "USD", "status": "pending"},
]

# DE Task 1: ids of failed or pending
bad_ids = [row["id"] for row in transactions if row["status"] != "success"]
print(bad_ids)
```

```python
# DE Task 2: user and amount for NGN rows
labels = [f"{row['user']}: {row['amount']}" for row in transactions if row["currency"] == "NGN"]
for label in labels:
    print(label)
```

```python
# DE Task 3: sort by user without changing the original
by_user = sorted(transactions, key=lambda row: row["user"])

for row in by_user:
    print(f"{row['user']}: {row['amount']}")

print()
print(f"original first row is still id {transactions[0]['id']}")
```

```python
# DE Task 4: split into big and small
big = [row for row in transactions if row["amount"] >= 1000]
small = [row for row in transactions if row["amount"] < 1000]

print(f"big: {len(big)}, small: {len(small)}")
```

```python
# DE Task 5: list of lists into list of dicts
raw = [[1, "ahmed", 5000], [2, "lahla", 120], [3, "judith", 900]]
columns = ["id", "user", "amount"]

rows = [dict(zip(columns, record)) for record in raw]

for row in rows:
    print(row)
```

## Tuples: tasks

```python
# Task 1: unpack a tuple
me = ("Ahmed", "Noisy-le-Grand", "Python")
name, city, language = me

print(name)
print(city)
print(language)
```

```python
# Task 2: min_max
def min_max(numbers):
    """Return the smallest and largest values as a tuple."""
    return min(numbers), max(numbers)

smallest, largest = min_max([4, 9, 15, 22, 7])
print(f"smallest={smallest} largest={largest}")
```

```python
# Task 3: zip into a dict
columns = ["month", "sales", "profit"]
values = ["JAN", 340, 120]

row = dict(zip(columns, values))
print(row)
```

```python
# DE Task 4: composite keys
sales = {
    ("Lagos", "JAN"): 340,
    ("Lagos", "FEB"): 318,
    ("Abuja", "JAN"): 205,
    ("Abuja", "FEB"): 260,
    ("Kano",  "JAN"): 150,
    ("Kano",  "FEB"): 175,
}

for (region, month), value in sales.items():
    print(f"{region} {month}: {value}")

print(f"total: {sum(sales.values())}")
```

```python
# DE Task 5: sort tuples by the second item
rows = [("ahmed", 5000), ("lahla", 120), ("judith", 900), ("uche", 4300)]

ranked = sorted(rows, key=lambda pair: pair[1], reverse=True)
print(ranked[:2])

for name, amount in ranked[:2]:
    print(f"{name}: {amount}")
```

## Sets: tasks

```python
# Task 1: unique values
numbers = [1, 2, 2, 3, 3, 3, 4]
unique = set(numbers)

print(unique)
print(f"{len(numbers) - len(unique)} duplicate(s) dropped")
```

```python
# Task 2: skill sets
mine = {"python", "sql", "excel"}
theirs = {"sql", "dbt", "excel"}

print(f"both:      {mine & theirs}")
print(f"mine only: {mine - theirs}")
print(f"combined:  {mine | theirs}")
```

```python
# DE Task 3: missing columns
expected = {"order_id", "customer_id", "amount", "order_date"}
received = {"order_id", "customer_id", "amount"}

missing = expected - received

if missing:
    print(f"PIPELINE FAILED: missing {len(missing)} column(s) -> {missing}")
else:
    print("Schema OK")
```

```python
# DE Task 4: load reconciliation
source = {1, 2, 3, 4, 5, 6, 7, 8}
loaded = {1, 2, 3, 5, 8}

missing = source - loaded
rate = len(loaded & source) / len(source) * 100

print(f"missing ids: {sorted(missing)}")
print(f"missing count: {len(missing)}")
print(f"completion rate: {round(rate, 1)}%")
```

```python
# DE Task 5: overlapping emails
list_a = ["a@x.com", "b@x.com", "a@x.com", "c@x.com"]
list_b = ["b@x.com", "c@x.com", "c@x.com", "d@x.com"]

set_a = set(list_a)
set_b = set(list_b)

print(f"in both: {set_a & set_b}")
print(f"unique across both: {len(set_a | set_b)}")
```

```python
transactions = [
    {"id": 1, "user": "ahmed",  "amount": 5000, "currency": "NGN", "status": "success"},
    {"id": 2, "user": "lahla",  "amount": 120,  "currency": "USD", "status": "failed"},
    {"id": 3, "user": "judith", "amount": 900,  "currency": "NGN", "status": "success"},
    {"id": 4, "user": "ahmed",  "amount": 4300, "currency": "NGN", "status": "success"},
    {"id": 5, "user": "uche",   "amount": 80,   "currency": "USD", "status": "pending"},
]

# DE Task 5: total per user, then the biggest
totals = {}
for row in transactions:
    totals[row["user"]] = totals.get(row["user"], 0) + row["amount"]

print(totals)

top_user = None
top_total = 0
for user, total in totals.items():
    if top_user is None or total > top_total:
        top_user = user
        top_total = total

print(f"top spender: {top_user} with {top_total}")
```

```python
# DE Task 6: group ids by status
grouped = {}
for row in transactions:
    grouped.setdefault(row["status"], []).append(row["id"])

print(grouped)
for status, ids in grouped.items():
    print(f"{status}: {len(ids)} -> {ids}")
```

## Mini project: full worked solution

```python
raw_orders = [
    {"order_id": 1, "customer": "ahmed",  "item": "laptop",  "amount": 450000, "country": "NG"},
    {"order_id": 2, "customer": "lahla",  "item": "phone",   "amount": 180000, "country": "GH"},
    {"order_id": 3, "customer": "ahmed",  "item": "mouse",   "amount": 8000,   "country": "NG"},
    {"order_id": 4, "customer": "judith", "item": "laptop",  "amount": 520000, "country": "NG"},
    {"order_id": 5, "customer": "uche",   "item": "monitor", "amount": 95000,  "country": "KE"},
    {"order_id": 6, "customer": "lahla",  "item": "mouse",   "amount": 7500,   "country": "GH"},
    {"order_id": 2, "customer": "lahla",  "item": "phone",   "amount": 180000, "country": "GH"},
    {"order_id": 7, "customer": "judith", "item": "keyboard","amount": 15000,  "country": "NG"},
]

# 1. deduplicate, keeping order
seen_ids = set()
orders = []
for order in raw_orders:
    if order["order_id"] in seen_ids:
        continue
    seen_ids.add(order["order_id"])
    orders.append(order)

print(f"in: {len(raw_orders)}, kept: {len(orders)}, dropped: {len(raw_orders) - len(orders)}")
```

```python
# 2. distinct values
countries = {order["country"] for order in orders}
items = {order["item"] for order in orders}

print(f"countries ({len(countries)}): {sorted(countries)}")
print(f"items ({len(items)}): {sorted(items)}")
```

```python
# 3. revenue per country
revenue_by_country = {}
for order in orders:
    revenue_by_country[order["country"]] = (
        revenue_by_country.get(order["country"], 0) + order["amount"]
    )

print(revenue_by_country)
```

```python
# 4. orders per customer
orders_by_customer = {}
for order in orders:
    orders_by_customer.setdefault(order["customer"], []).append(order["order_id"])

print(orders_by_customer)
```

```python
# 5. top spender as a tuple
spend = {}
for order in orders:
    spend[order["customer"]] = spend.get(order["customer"], 0) + order["amount"]

top = max(spend.items(), key=lambda pair: pair[1])
print(top)

name, total = top
print(f"{name} spent {total}")
```

```python
# 6. composite key: (country, item) -> total
by_country_item = {}
for order in orders:
    key = (order["country"], order["item"])
    by_country_item[key] = by_country_item.get(key, 0) + order["amount"]

for key, total in sorted(by_country_item.items(), key=lambda pair: pair[1], reverse=True):
    country, item = key
    print(f"{country} / {item}: {total}")
```

```python
# 7. big orders
big_orders = sorted(
    [order for order in orders if order["amount"] > 100_000],
    key=lambda order: order["amount"],
    reverse=True,
)

for order in big_orders:
    print(f"id {order['order_id']}: {order['customer']} {order['amount']}")
```

```python
# 8. the report
print("=== WEEKLY ORDERS REPORT ===")
print(f"Raw rows:       {len(raw_orders)}")
print(f"After dedupe:   {len(orders)}")
print(f"Duplicates:     {len(raw_orders) - len(orders)}")
print()
print(f"Countries:      {sorted(countries)}")
print(f"Items:          {sorted(items)}")
print()
print("Revenue by country:")
for country, total in sorted(revenue_by_country.items(), key=lambda p: p[1], reverse=True):
    print(f"   {country}: {total:,}")
print()
print("Orders per customer:")
for customer, ids in orders_by_customer.items():
    print(f"   {customer}: {len(ids)} order(s) {ids}")
print()
print(f"Top spender:    {name} ({total:,})")
print(f"Big orders:     {len(big_orders)} over 100,000")
print(f"Total revenue:  {sum(order['amount'] for order in orders):,}")
```

```python
# Stretch 9: reconcile against finance
finance_ids = {1, 2, 3, 4, 5, 9}
our_ids = {order["order_id"] for order in orders}

print(f"we have that finance is missing: {our_ids - finance_ids}")
print(f"finance has that we have never seen: {finance_ids - our_ids}")
print(f"agreed on: {sorted(our_ids & finance_ids)}")
```

```python
# Stretch 10: a customer -> country lookup
customer_country = {order["customer"]: order["country"] for order in orders}
print(customer_country)

order_4 = {order["order_id"]: order for order in orders}[4]
print(f"order 4 is from {customer_country[order_4['customer']]}")
```

```python
# Stretch 11: average order value per country
count_by_country = {}
for order in orders:
    count_by_country[order["country"]] = count_by_country.get(order["country"], 0) + 1

for country, total in revenue_by_country.items():
    average = total / count_by_country[country]
    print(f"{country}: {count_by_country[country]} order(s), average {round(average, 2):,}")
```
