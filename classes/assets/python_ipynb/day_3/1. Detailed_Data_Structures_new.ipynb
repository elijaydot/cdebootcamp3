### Slicing, properly

`mylist[start:stop:step]`. The `start` is included, the `stop` is **not**.
Leave a part out and Python fills in a sensible default.

````python
numbers = [10, 20, 30, 40, 50, 60, 70]

print(numbers[2:5])      # [30, 40, 50]   index 2, 3, 4. NOT 5
print(numbers[:3])       # first three
print(numbers[4:])       # from index 4 to the end
print(numbers[-3:])      # last three
print(numbers[:-1])      # everything except the last
print(numbers[::2])      # every second item
print(numbers[::-1])     # the whole list, reversed
print(numbers[:])        # a full copy
````

Slicing never crashes. `numbers[2:500]` gives you what exists and says nothing.
`numbers[500]` raises `IndexError`. That difference catches people out.

````python
numbers = [10, 20, 30]

print(numbers[1:500])       # fine, gives [20, 30]
print(numbers[5:9])         # fine, gives []

try:
    print(numbers[500])     # not fine
except IndexError as e:
    print(f"IndexError: {e}")
````

### The copy trap

This is the single most expensive lesson in Python lists. Assigning a list to a new
name does **not** copy it. Both names point at the same list in memory.

````python
original = ["apple", "banana", "orange"]
backup = original              # NOT a copy

backup.append("mango")

print(f"backup:   {backup}")
print(f"original: {original}")      # it changed too
print(f"same object? {original is backup}")
````

````python
# Three ways to make a real copy
original = ["apple", "banana", "orange"]

copy1 = original[:]            # slice
copy2 = list(original)         # constructor
copy3 = original.copy()        # method

copy1.append("mango")

print(f"copy1:    {copy1}")
print(f"original: {original}")      # untouched
print(f"same object? {original is copy1}")
````

In a pipeline this shows up as "why did my raw data change after I cleaned it".
When you write a function that modifies a list, copy first unless you deliberately
mean to change the caller's data.

````python
# append vs extend, side by side. You saw both above. Here is the difference in one cell.

a = ["orange", "grape"]
b = ["orange", "grape"]

a.append(["lemon", "mango"])     # adds ONE item, which happens to be a list
b.extend(["lemon", "mango"])     # adds EACH item separately

print(f"append: {a}    length {len(a)}")
print(f"extend: {b}    length {len(b)}")
````

Rule of thumb: `append` adds one thing. `extend` merges another list in.
`+` also merges, but builds a whole new list rather than changing the old one.

````python
a = [1, 2]
b = [3, 4]

print(a + b)          # new list, a is untouched
print(a)

a += b                # same as a.extend(b), changes a in place
print(a)
````

````python
# sort vs sorted: one changes the list, the other returns a new one

fruits = ["orange", "apple", "mango", "banana"]

new_list = sorted(fruits)        # returns a NEW sorted list
print(f"sorted() gave: {new_list}")
print(f"original still: {fruits}")

fruits.sort()                    # changes the list, returns None
print(f"after .sort(): {fruits}")

print()
print(f"careful: {fruits.sort()}")     # None. .sort() returns nothing
````

`x = mylist.sort()` is a classic bug. `x` becomes `None`, because `.sort()` sorts in
place and returns nothing. Use `sorted()` when you want a value back.

````python
# sorting with a key, and in reverse

words = ["python", "sql", "dbt", "airflow"]
print(sorted(words, key=len))                  # shortest first
print(sorted(words, key=len, reverse=True))    # longest first

rows = [
    {"user": "ahmed",  "amount": 5000},
    {"user": "lahla",  "amount": 120},
    {"user": "judith", "amount": 900},
]

by_amount = sorted(rows, key=lambda row: row["amount"], reverse=True)
for row in by_amount:
    print(row)
````

`key=lambda row: row["amount"]` says "when comparing two rows, compare their amounts".
A `lambda` is just a tiny unnamed function. You will meet it constantly in pandas.

````python
# Built-in functions that work on any list of numbers

sales = [120, 450, 800, 50, 900, 300]

print(f"count:   {len(sales)}")
print(f"total:   {sum(sales)}")
print(f"lowest:  {min(sales)}")
print(f"highest: {max(sales)}")
print(f"average: {sum(sales) / len(sales)}")
print(f"rounded: {round(sum(sales) / len(sales), 2)}")
````

### List comprehensions

A loop that builds a list, written on one line. You will see these everywhere, and
pandas will feel much easier once you can read them.

```
[ what_to_keep  for item in collection  if condition ]
```

````python
sales = [120, 450, 800, 50, 900, 300]

# The long way
doubled = []
for amount in sales:
    doubled.append(amount * 2)

# The comprehension way, exactly the same result
doubled_short = [amount * 2 for amount in sales]

print(doubled)
print(doubled_short)
````

````python
# With a filter on the end
big_sales = [amount for amount in sales if amount > 400]
print(big_sales)

# Pulling one field out of a list of dicts. You will do this constantly.
rows = [
    {"user": "ahmed",  "amount": 5000},
    {"user": "lahla",  "amount": 120},
    {"user": "judith", "amount": 900},
]

users = [row["user"] for row in rows]
amounts = [row["amount"] for row in rows]
big_users = [row["user"] for row in rows if row["amount"] > 500]

print(users)
print(amounts)
print(big_users)
````

**When not to use one.** If it no longer fits comfortably on one line, or it has two
`if`s and a nested loop inside it, write the normal loop. Clever is not the goal.
Readable is.

````python
# Other things worth knowing

items = ["a", "b", "c", "b", "d"]

print("b" in items)              # membership
print(items.index("b"))          # position of the FIRST b
print(items.count("b"))          # how many b
print(len(items))

del items[0]                     # delete by position
print(items)

items.clear()                    # empty it
print(items)
````

````python
# Nested lists: a list of lists is a table of rows

grid = [
    ["JAN", 340, 360],
    ["FEB", 318, 342],
    ["MAR", 362, 406],
]

print(grid[0])          # the first row
print(grid[0][1])       # first row, second column
print(grid[1][0])       # FEB

for row in grid:
    print(f"{row[0]}: {row[1]} then {row[2]}")
````

### Real life: lists in a data pipeline

The idea that makes all of this click:

- **One row of a table is a dictionary.** The column names are the keys.
- **A whole table is a list of those dictionaries.**

Every dataset you ever load, before it becomes a DataFrame, looks like this.

````python
transactions = [
    {"id": 1, "user": "ahmed",  "amount": 5000,  "currency": "NGN", "status": "success"},
    {"id": 2, "user": "lahla",  "amount": 120,   "currency": "USD", "status": "failed"},
    {"id": 3, "user": "judith", "amount": 900,   "currency": "NGN", "status": "success"},
    {"id": 4, "user": "uche",   "amount": 4300,  "currency": "NGN", "status": "success"},
    {"id": 5, "user": "ahmed",  "amount": 80,    "currency": "USD", "status": "pending"},
]

print(f"{len(transactions)} rows")
print(transactions[0])                 # first row
print(transactions[0]["user"])         # one field of one row
````

````python
# DE example 1: pulling a single column out

amounts = [row["amount"] for row in transactions]
print(amounts)
print(f"total: {sum(amounts)}, biggest: {max(amounts)}")

# and only the successful ones
successful_amounts = [row["amount"] for row in transactions if row["status"] == "success"]
print(f"successful total: {sum(successful_amounts)}")
````

````python
# DE example 2: batching, using slices
# You almost never load 10 million rows in one go. You send them in chunks.

rows = list(range(1, 24))          # pretend these are 23 rows
batch_size = 5

start = 0
batch_number = 0
while start < len(rows):
    batch = rows[start:start + batch_size]
    batch_number += 1
    print(f"batch {batch_number}: {len(batch)} rows -> {batch}")
    start += batch_size
````

````python
# DE example 3: dedupe while keeping the original order

user_ids = [101, 102, 101, 103, 102, 104, 101]

seen = set()
unique = []
for uid in user_ids:
    if uid not in seen:
        seen.add(uid)
        unique.append(uid)

print(f"in: {len(user_ids)}, out: {len(unique)}, dropped: {len(user_ids) - len(unique)}")
print(unique)
````

````python
# DE example 4: sorting a table by a column, then taking the top N

top_3 = sorted(transactions, key=lambda row: row["amount"], reverse=True)[:3]

print("Top 3 transactions by amount:")
for row in top_3:
    print(f"  id {row['id']}: {row['user']} {row['amount']} {row['currency']}")
````

That one line is `ORDER BY amount DESC LIMIT 3`. Sort, then slice.

## Tuples
Tuples are ordered, immutable, and allow duplicate elements. Since they are immutable, their contents cannot be modified after creation.
### Tuple Operations

**A tuple:**

- Is ordered
- Cannot be changed (immutable)
- Allows duplicates

````python
# Create a tuple
my_tuple = (1, 2, 3, 4, 5)

# Access elements by index
print(my_tuple[0])      # First element
print(my_tuple[-1])     # Last element

# Slicing
print(my_tuple[1:4])    # Slice from index 1 to 3
````

### Common Tuple Methods

````python
# Tuples have fewer methods than lists due to their immutability.
# Two common methods are `count` and `index`.

# Count: Count the number of occurrences of a value
print("Count of 3:", my_tuple.count(3))

# Index: Find the index of the first occurrence of a value
print("Index of 4:", my_tuple.index(4))
````

## Practice: Tuples

### Why would anyone want a list that cannot change?

Three good reasons, and they are all about safety.

1. Nobody can modify it by accident. A config, a set of column names, a coordinate.
2. It can be used as a **dictionary key**. A list cannot.
3. It signals intent: "these belong together and the order means something".

````python
point = (6.5244, 3.3792)        # latitude, longitude for Lagos

print(point[0])
print(len(point))

try:
    point[0] = 99
except TypeError as e:
    print(f"TypeError: {e}")
````

````python
# Unpacking: give each position its own name
point = (6.5244, 3.3792)
latitude, longitude = point
print(f"lat {latitude}, lon {longitude}")

# This is why functions can return several things at once
def get_stats(numbers):
    return len(numbers), sum(numbers), max(numbers)

count, total, biggest = get_stats([4, 9, 15])
print(count, total, biggest)
````

````python
# Swapping without a temporary variable. Tuples make this work.
a = 1
b = 2
a, b = b, a
print(a, b)

# Ignoring a value you do not need, by convention with an underscore
row = (101, "ahmed", 5000, "NGN")
user_id, name, _, currency = row
print(user_id, name, currency)
````

### The one-item tuple trap

````python
not_a_tuple = ("hello")          # just a string in brackets
actually_a_tuple = ("hello",)    # the comma is what makes it a tuple

print(type(not_a_tuple).__name__)
print(type(actually_a_tuple).__name__)
print(len(actually_a_tuple))
````

````python
# Tuples as dictionary keys, which lists cannot do

sales_by_region_month = {
    ("Lagos", "JAN"): 340,
    ("Lagos", "FEB"): 318,
    ("Abuja", "JAN"): 205,
}

print(sales_by_region_month[("Lagos", "JAN")])

for (region, month), value in sales_by_region_month.items():
    print(f"{region} {month}: {value}")

# A list as a key fails
try:
    broken = {["Lagos", "JAN"]: 340}
except TypeError as e:
    print(f"TypeError: {e}")
````

That composite key is exactly what a `GROUP BY region, month` produces. Tuples are
how Python represents "a combination of columns".

````python
# zip() pairs two lists into tuples. Extremely useful.

columns = ["id", "user", "amount"]
values = [1, "ahmed", 5000]

print(list(zip(columns, values)))
print(dict(zip(columns, values)))       # instant row dictionary

for column, value in zip(columns, values):
    print(f"{column} = {value}")
````

### List or tuple? A decision you make constantly

| Use a **list** when | Use a **tuple** when |
|---|---|
| The contents will change | The contents are fixed |
| It is a collection of similar things (rows, names) | It is one record with meaningful positions |
| You need `append`, `sort`, `remove` | You want it safe from modification |
| Order can be rearranged | You will use it as a dict key |

## Sets
Sets are unordered collections of unique elements. They are used to perform operations like union, intersection, and difference.
### Set Operations and Methods

**A set:**

- Is unordered
- Cannot have duplicates
- Is changeable

````python
my_set = {1, 2, 3, 4, 5}
````

````python
my_set
````

````python
# Create a set
my_set = {1, 2, 3, 4, 5}

# Add: Add an element to the set
my_set.add(6)
print(my_set)

# Remove: Remove an element (raises KeyError if element is not found)
my_set.remove(3)
print(my_set)

# Discard: Remove an element (does not raise an error if element is not found)
my_set.discard(16)   # No error
print(my_set)

# Pop: Remove and return a random element
random_item = my_set.pop()
print("Popped item:", random_item)
print(my_set)

# Clear: Remove all elements from the set
my_set.clear()
print("Cleared set:", my_set)
````

### Set Operations

````python
set_a = {1, 2, 3}
set_b = {3, 4, 5}

# Union: Combine two sets (all unique elements)
print("Union:", set_a.union(set_b))

# Intersection: Elements present in both sets
print("Intersection:", set_a.intersection(set_b))

# Difference: Elements in set_a but not in set_b
print("Difference:", set_a.difference(set_b))

# Symmetric Difference: Elements in either set, but not both
print("Symmetric Difference:", set_a.symmetric_difference(set_b))
````

## Practice: Sets

### The operators, which are shorter than the method names

````python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

print(f"union        {set_a | set_b}      same as set_a.union(set_b)")
print(f"intersection {set_a & set_b}            .intersection()")
print(f"difference   {set_a - set_b}            .difference()")
print(f"in b not a   {set_b - set_a}")
print(f"symmetric    {set_a ^ set_b}      .symmetric_difference()")
````

Note that `a - b` and `b - a` give different answers. Difference is not symmetric.
That asymmetry is exactly what makes it useful for reconciliation: "what is in the
source but not the target" is a different question from the reverse.

````python
# Subset and superset: is everything in A also in B?

required = {"id", "email"}
arrived  = {"id", "email", "phone", "country"}

print(required.issubset(arrived))      # do we have everything we need?
print(arrived.issuperset(required))    # same question, other way round
print(required <= arrived)             # the operator version

print({1, 2} & {3, 4})                 # empty set
print({1, 2}.isdisjoint({3, 4}))       # no overlap at all
````

````python
# Sets drop duplicates the moment you create them

user_ids = [101, 102, 101, 103, 102, 104, 101]
unique = set(user_ids)

print(unique)
print(f"{len(user_ids)} rows in, {len(unique)} unique users")
print(f"{len(user_ids) - len(unique)} duplicate row(s)")
````

**But you lose the order.** A set has no first item and no last item. If the order
matters, use the `seen` pattern from the Lists section instead.

````python
# Why sets are not just "lists without duplicates": membership is fast

big_list = list(range(200_000))
big_set = set(big_list)

# Both of these answer the same question
print(199_999 in big_list)
print(199_999 in big_set)

# The list checks every item one by one. The set jumps straight there.
# At 200k items you can feel it. At 10 million it decides whether your job finishes.
import time

start = time.perf_counter()
199_999 in big_list
list_time = time.perf_counter() - start

start = time.perf_counter()
199_999 in big_set
set_time = time.perf_counter() - start

print(f"list lookup: {list_time * 1000:.4f} ms")
print(f"set lookup:  {set_time * 1000:.4f} ms")
````

````python
# Adding and removing: remove raises, discard does not

tables = {"customers", "orders"}

tables.add("payments")
print(tables)

tables.discard("refunds")        # not there, no complaint
print(tables)

try:
    tables.remove("refunds")     # not there, raises
except KeyError as e:
    print(f"KeyError: {e}")
````

### Real life: sets in a data pipeline

Sets answer four questions you will be asked every week.

````python
# DE example 1: schema drift. Did the columns we expect actually arrive?

expected = {"id", "email", "created_at", "country"}
received = {"id", "email", "country", "phone"}

missing = expected - received
unexpected = received - expected

if missing:
    print(f"FAIL: missing columns {missing}")
if unexpected:
    print(f"WARN: new columns appeared {unexpected}")
if not missing and not unexpected:
    print("Schema matches exactly")
````

````python
# DE example 2: reconciliation. Did every row make it from source to target?

source_ids = {101, 102, 103, 104, 105, 106}
target_ids = {101, 102, 104, 106, 999}

not_loaded = source_ids - target_ids
extra_in_target = target_ids - source_ids

print(f"in source: {len(source_ids)}, in target: {len(target_ids)}")
print(f"failed to load: {not_loaded}")
print(f"in target but not source (suspicious): {extra_in_target}")
print(f"matched: {len(source_ids & target_ids)}")
````

````python
# DE example 3: which customers appear in both systems?

crm_emails = {"a@x.com", "b@x.com", "c@x.com"}
billing_emails = {"b@x.com", "c@x.com", "d@x.com"}

print(f"in both:        {crm_emails & billing_emails}")
print(f"crm only:       {crm_emails - billing_emails}")
print(f"billing only:   {billing_emails - crm_emails}")
print(f"everyone:       {crm_emails | billing_emails}")
print(f"only one system: {crm_emails ^ billing_emails}")
````

````python
# DE example 4: a fast allow-list check inside a loop

allowed_statuses = {"success", "failed", "pending"}

rows = [
    {"id": 1, "status": "success"},
    {"id": 2, "status": "reversed"},
    {"id": 3, "status": "pending"},
]

for row in rows:
    if row["status"] not in allowed_statuses:
        print(f"id {row['id']}: unexpected status {row['status']!r}")
    else:
        print(f"id {row['id']}: ok")
````

## Dictionaries
Dictionaries are collections of key-value pairs. They are unordered, mutable, and allow for fast lookups by key.
### Dictionary Operations and Methods

**A dictionary:**

- Stores data as key-value pairs
- Is ordered (Python 3.7+)
- Is changeable

````python
# Create a dictionary
my_dict = {'a': 1, 'b': 2, 'c': 3}

# Access a value by key
print(my_dict['a'])

# Add or update a key-value pair
my_dict['d'] = 4
print(my_dict)

# Remove a key-value pair
removed_value = my_dict.pop('b')
print("Removed value:", removed_value)
print(my_dict)

# Get: Access a value safely (returns None if key is not found)
value = my_dict.get('e', 'Not found')
print("Value for key 'e':", value)

# Keys, values, and items
print("Keys:", my_dict.keys())
print("Values:", my_dict.values())
print("Items:", my_dict.items())
````

### Reading safely: `[ ]` versus `.get()`

````python
row = {"id": 1, "user": "ahmed", "amount": 5000}

print(row["user"])                   # crashes if the key is missing
print(row.get("user"))               # returns None if missing
print(row.get("email"))              # None, no crash
print(row.get("email", "unknown"))   # your own fallback

try:
    print(row["email"])
except KeyError as e:
    print(f"KeyError: {e}")
````

Use `[ ]` when the key **must** be there and a missing one is a real error you want
to hear about. Use `.get()` when it is genuinely optional. Both are correct in
different places, and choosing deliberately is what separates careful code from
code that crashes at 2am.

````python
# Adding, updating, removing

student = {"name": "John", "age": 22}

student["course"] = "Data Science"       # add
student["age"] = 23                      # update, same syntax
print(student)

removed = student.pop("age")             # remove and give it back
print(f"removed {removed}, left with {student}")

student.pop("nickname", None)            # missing key, no crash, thanks to the default
print(student)

student.update({"level": 300, "course": "Data Engineering"})   # add and update at once
print(student)
````

````python
# The three ways to loop, and when each is right

prices = {"rice": 1200, "beans": 900, "garri": 500}

for key in prices:                       # keys only, the default
    print(key)

print()
for value in prices.values():            # values only
    print(value)

print()
for key, value in prices.items():        # both, which is usually what you want
    print(f"{key}: {value}")
````

````python
# The counting pattern. Memorise this one.

statuses = ["success", "failed", "success", "pending", "success", "failed"]

counts = {}
for status in statuses:
    if status in counts:
        counts[status] += 1
    else:
        counts[status] = 1

print(counts)

# Shorter, using .get with a default of 0
counts2 = {}
for status in statuses:
    counts2[status] = counts2.get(status, 0) + 1

print(counts2)
````

````python
# The grouping pattern: a dictionary whose values are lists

rows = [
    {"user": "ahmed",  "amount": 5000},
    {"user": "lahla",  "amount": 120},
    {"user": "ahmed",  "amount": 900},
    {"user": "judith", "amount": 4300},
    {"user": "lahla",  "amount": 80},
]

grouped = {}
for row in rows:
    user = row["user"]
    if user not in grouped:
        grouped[user] = []
    grouped[user].append(row["amount"])

print(grouped)

for user, amounts in grouped.items():
    print(f"{user}: {len(amounts)} transaction(s), total {sum(amounts)}")
````

That is `GROUP BY user` with the rows kept. The counting pattern above is
`GROUP BY status` with only the count. You now have both.

````python
# setdefault does the "if not in, create it" dance in one line

grouped = {}
for row in rows:
    grouped.setdefault(row["user"], []).append(row["amount"])

print(grouped)
````

````python
# Dictionary comprehensions, the same idea as list comprehensions

prices = {"rice": 1200, "beans": 900, "garri": 500}

doubled = {item: price * 2 for item, price in prices.items()}
print(doubled)

cheap = {item: price for item, price in prices.items() if price < 1000}
print(cheap)

# Building a lookup table from a list of rows. Very common.
users = [
    {"id": 101, "name": "ahmed"},
    {"id": 102, "name": "lahla"},
    {"id": 103, "name": "judith"},
]

by_id = {row["id"]: row["name"] for row in users}
print(by_id)
print(by_id[102])
````

That lookup table is a **join**, done by hand. Instead of scanning a list for every
id, you build the dictionary once and every lookup after that is instant.

````python
# Nested dictionaries: what every API response looks like

response = {
    "page": 1,
    "meta": {"total": 250, "per_page": 50},
    "data": [
        {"id": 1, "user": {"name": "ahmed", "country": "NG"}, "amount": 5000},
        {"id": 2, "user": {"name": "lahla", "country": "GH"}, "amount": 120},
    ],
}

print(response["meta"]["total"])
print(response["data"][0]["user"]["name"])

for record in response["data"]:
    name = record["user"]["name"]
    country = record["user"]["country"]
    print(f"{record['id']}: {name} ({country}) {record['amount']}")
````

````python
# Reaching into a nested dict safely, when parts might be missing

record = {"id": 3, "amount": 700}          # no "user" key at all

# This crashes
try:
    print(record["user"]["name"])
except KeyError as e:
    print(f"KeyError: {e}")

# This does not
name = record.get("user", {}).get("name", "unknown")
print(name)
````

`record.get("user", {})` returns an empty dictionary when `user` is missing, and
calling `.get("name")` on an empty dictionary is perfectly safe. Chain it and your
parser stops crashing on incomplete records.

````python
# Merging dictionaries: config defaults overridden by what the user supplied

defaults = {"host": "localhost", "port": 5432, "retries": 3}
supplied = {"host": "prod-db", "retries": 5}

final = {**defaults, **supplied}      # later one wins
print(final)

final2 = defaults | supplied          # same thing, Python 3.9+
print(final2)
````

### Real life: dictionaries in a data pipeline

````python
transactions = [
    {"id": 1, "user": "ahmed",  "amount": 5000, "currency": "NGN", "status": "success"},
    {"id": 2, "user": "lahla",  "amount": 120,  "currency": "USD", "status": "failed"},
    {"id": 3, "user": "judith", "amount": 900,  "currency": "NGN", "status": "success"},
    {"id": 4, "user": "ahmed",  "amount": 4300, "currency": "NGN", "status": "success"},
    {"id": 5, "user": "uche",   "amount": 80,   "currency": "USD", "status": "pending"},
]

# DE example 1: SUM(amount) GROUP BY currency
totals = {}
for row in transactions:
    currency = row["currency"]
    totals[currency] = totals.get(currency, 0) + row["amount"]

print(totals)
````

````python
# DE example 2: a config dictionary driving the pipeline
# Change the config, not the code. This is how real jobs are built.

pipeline_config = {
    "source": "postgres",
    "tables": ["customers", "orders", "payments"],
    "batch_size": 5000,
    "checks": {"null_check": True, "row_count_check": True, "schema_check": False},
}

print(f"Loading {len(pipeline_config['tables'])} table(s) from {pipeline_config['source']}")

for table in pipeline_config["tables"]:
    print(f"  {table} in batches of {pipeline_config['batch_size']}")

for check, enabled in pipeline_config["checks"].items():
    if enabled:
        print(f"  running {check}")
    else:
        print(f"  skipping {check}")
````

````python
# DE example 3: a lookup table instead of a nested loop
# The slow way scans the whole users list once per transaction.
# The fast way builds the lookup once, then jumps straight to the answer.

users = [
    {"id": 101, "name": "ahmed",  "country": "NG"},
    {"id": 102, "name": "lahla",  "country": "GH"},
    {"id": 103, "name": "judith", "country": "NG"},
]

orders = [
    {"order_id": 1, "user_id": 101, "amount": 5000},
    {"order_id": 2, "user_id": 103, "amount": 900},
    {"order_id": 3, "user_id": 999, "amount": 50},      # unknown user
]

user_lookup = {user["id"]: user for user in users}

for order in orders:
    user = user_lookup.get(order["user_id"])
    if user is None:
        print(f"order {order['order_id']}: ORPHAN, user {order['user_id']} not found")
    else:
        print(f"order {order['order_id']}: {user['name']} ({user['country']}) {order['amount']}")
````

That is a `LEFT JOIN`, and the orphan row is exactly what a left join would leave
with nulls. Real pipelines check for these, because an orphan usually means the
two tables were extracted at different times.

````python
# DE example 4: turning rows into a summary dictionary you can print or store

summary = {
    "row_count": len(transactions),
    "by_status": {},
    "by_currency": {},
    "distinct_users": len({row["user"] for row in transactions}),
}

for row in transactions:
    summary["by_status"][row["status"]] = summary["by_status"].get(row["status"], 0) + 1
    summary["by_currency"][row["currency"]] = (
        summary["by_currency"].get(row["currency"], 0) + row["amount"]
    )

print(summary)
print()
for key, value in summary.items():
    print(f"{key}: {value}")
````

## Choosing a structure: the one table to remember

| | List | Tuple | Set | Dictionary |
|---|---|---|---|---|
| Written with | `[ ]` | `( )` | `{ }` | `{key: value}` |
| Ordered | yes | yes | no | yes (3.7+) |
| Changeable | yes | no | yes | yes |
| Duplicates | yes | yes | no | keys no, values yes |
| Access by | position | position | membership only | key |
| Fast lookup | no | no | yes | yes |
| Can be a dict key | no | yes | no | no |

**Ask yourself one question at a time:**

1. Do I need to look things up by name? → **dictionary**
2. Do I only care whether something is present, or need uniqueness? → **set**
3. Will it never change, or do I need it as a dict key? → **tuple**
4. Otherwise → **list**

In data engineering terms:

- A **row** is a dictionary
- A **table** is a list of dictionaries
- A **column** is a list you pull out of that table
- **Distinct values, schema checks and reconciliation** are sets
- A **composite key** or a fixed record is a tuple

## Common mistakes to watch for

| Mistake | What you see | The fix |
|---|---|---|
| `backup = original` for a list | Both change together | `original.copy()` or `original[:]` |
| `x = mylist.sort()` | `x` is `None` | `.sort()` changes in place, `sorted()` returns |
| `mylist.append(other_list)` | A list inside your list | Use `.extend()` to merge |
| `row["missing_key"]` | `KeyError` | `.get()` with a default, or check `in` first |
| `mylist[500]` | `IndexError` | Check `len()` first, or slice instead |
| `("hello")` meant as a tuple | It is a string | `("hello",)` with the comma |
| `{["a"]: 1}` | `TypeError: unhashable` | Dict keys must be immutable, use a tuple |
| Expecting order from a set | Items come back shuffled | Use a list, or the `seen` pattern |
| `counts[key] += 1` on a new key | `KeyError` | `counts.get(key, 0) + 1` |
| Modifying a list while looping it | Items get skipped | Build a new list instead |

````python
# See the last one for yourself. This does NOT remove every odd number.

numbers = [1, 2, 3, 4, 5, 6, 7]
for n in numbers:
    if n % 2 == 1:
        numbers.remove(n)

print(f"wrong: {numbers}")

# Build a new list instead
numbers = [1, 2, 3, 4, 5, 6, 7]
evens = [n for n in numbers if n % 2 == 0]
print(f"right: {evens}")
````
