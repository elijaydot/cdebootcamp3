## Functions: beginner tasks

```python
# Task 1: square
def square(n):
    """Return n multiplied by itself."""
    return n * n

print(square(7))
```

```python
# Task 2: is_even
def is_even(n):
    """Return True when n is even."""
    return n % 2 == 0

for number in [1, 2, 3, 4, 5, 6]:
    if is_even(number):
        print(number)
```

```python
# Task 3: greet returns instead of printing
def greet(name, greeting="Hello"):
    """Return a greeting string."""
    return f"{greeting}, {name}!"

print(greet("Ahmed"))
print(greet("Ahmed", "Sannu"))
```

```python
# Task 4: longest word
def longest_word(words):
    """Return the longest word in a list."""
    longest = words[0]
    for word in words:
        if len(word) > len(longest):
            longest = word
    return longest

print(longest_word(["python", "sql", "airflow", "dbt"]))
```

```python
# Task 5: stats returns three values
def stats(numbers):
    """Return the count, total and average of a list of numbers."""
    total = 0
    for n in numbers:
        total = total + n
    return len(numbers), total, total / len(numbers)

count, total, average = stats([4, 9, 15, 22, 7])
print(f"count={count} total={total} average={average}")
```

```python
# Task 6: shout with *args
def shout(*words):
    """Return all the words joined with spaces, in capitals."""
    return " ".join(words).upper()

print(shout("data", "is", "power"))

# and with a list, using the star at the call site
parts = ["dec", "cohort", "2026"]
print(shout(*parts))
```

## Functions: DE tasks

```python
transactions = [
    {"id": 1, "user": "ahmed",  "amount": 5000, "currency": "NGN", "status": "success"},
    {"id": 2, "user": "lahla",  "amount": 900,  "currency": "NGN", "status": "failed"},
    {"id": 3, "user": "judith", "amount": 120,  "currency": "USD", "status": "success"},
    {"id": 4, "user": "uche",   "amount": 800,  "currency": "NGN", "status": "success"},
]

# DE Task 1: email check
def is_valid_email(email):
    """True when the string has an @ with a dot somewhere after it."""
    if "@" not in email:
        return False
    name, _, domain = email.partition("@")
    return name != "" and "." in domain

for test in ["a@x.com", "bad-email", "no@dotcom", ""]:
    print(f"{test!r:>14} -> {is_valid_email(test)}")
```

```python
# DE Task 2: count_by
def count_by(rows, column):
    """Count how many rows have each distinct value of a column."""
    counts = {}
    for row in rows:
        key = row.get(column)
        if key in counts:
            counts[key] += 1
        else:
            counts[key] = 1
    return counts

print(count_by(transactions, "status"))
print(count_by(transactions, "currency"))
```

```python
# DE Task 3: filter_rows
def filter_rows(rows, column, value):
    """Return only the rows where column equals value."""
    result = []
    for row in rows:
        if row.get(column) == value:
            result.append(row)
    return result

successes = filter_rows(transactions, "status", "success")
print(f"{len(successes)} successful row(s)")
for row in successes:
    print(" ", row["id"], row["user"])
```

```python
# DE Task 4: add_column without touching the originals
def add_column(rows, name, value):
    """Return a new list of rows, each with an extra key."""
    result = []
    for row in rows:
        new_row = dict(row)        # a copy, so the original is untouched
        new_row[name] = value
        result.append(new_row)
    return result

stamped = add_column(transactions, "loaded_by", "dec_pipeline")
print(stamped[0])
print(transactions[0])             # unchanged, which is the point
```

```python
# DE Task 5: summarise a numeric column
def summarise(rows, column):
    """Return the minimum, maximum and average of a numeric column."""
    values = []
    for row in rows:
        values.append(row[column])

    smallest = values[0]
    largest = values[0]
    total = 0
    for v in values:
        if v < smallest:
            smallest = v
        if v > largest:
            largest = v
        total = total + v

    return smallest, largest, total / len(values)

low, high, avg = summarise(transactions, "amount")
print(f"min={low} max={high} avg={avg}")
```

```python
# DE Task 6: build an INSERT statement from a dict
def build_insert(table, row):
    """Turn one row dict into an INSERT statement."""
    columns = []
    values = []
    for key, value in row.items():
        columns.append(key)
        if isinstance(value, str):
            values.append(f"'{value}'")
        else:
            values.append(str(value))
    return f"INSERT INTO {table} ({', '.join(columns)}) VALUES ({', '.join(values)});"

print(build_insert("payments", transactions[0]))
```

## Errors: DE tasks

```python
# DE Task 1: safe_get
def safe_get(row, column, default=None):
    """Return row[column], or the default when the column is missing."""
    try:
        return row[column]
    except KeyError:
        return default

row = {"id": 1, "user": "ahmed"}
print(safe_get(row, "user"))
print(safe_get(row, "amount"))
print(safe_get(row, "amount", 0))
```

```python
# DE Task 2: clean and rejected with try/except
raw = [
    {"id": 1, "email": "a@x.com", "amount": "500"},
    {"id": 2, "email": "b@x.com", "amount": "N/A"},
    {"id": 3,                     "amount": "700"},
    {"id": 4, "email": "d@x.com", "amount": None},
]

clean = []
rejected = []

for row in raw:
    try:
        record = {
            "id": row["id"],
            "email": row["email"],
            "amount": float(row["amount"]),
        }
        clean.append(record)
    except KeyError as e:
        rejected.append({"id": row["id"], "reason": f"missing column {e}"})
    except (ValueError, TypeError) as e:
        rejected.append({"id": row["id"], "reason": f"bad amount: {e}"})

print(f"clean: {len(clean)}, rejected: {len(rejected)}")
for bad in rejected:
    print(f"  id {bad['id']}: {bad['reason']}")
```

```python
# DE Task 3: a rate that survives an empty table
def divide_safely(numerator, denominator):
    """Return the ratio, or 0 when the denominator is zero."""
    try:
        return numerator / denominator
    except ZeroDivisionError:
        return 0

empty_table_rows = 0
nulls_found = 0

null_rate = divide_safely(nulls_found, empty_table_rows) * 100
print(f"null rate: {null_rate}%")
print(f"normal case: {divide_safely(3, 12) * 100}%")
```

```python
# DE Task 4: validate a partition date
def load_partition(date_string):
    """Raise ValueError unless the string looks like YYYY-MM-DD."""
    if len(date_string) != 10:
        raise ValueError(f"expected YYYY-MM-DD, got {date_string!r}")
    if date_string[4] != "-" or date_string[7] != "-":
        raise ValueError(f"expected dashes at positions 5 and 8, got {date_string!r}")
    return f"Loading partition {date_string}"

for candidate in ["2026-09-12", "12/09/2026"]:
    try:
        print(load_partition(candidate))
    except ValueError as e:
        print(f"REFUSED: {e}")
```

```python
# DE Task 5: retry with doubling wait
def flaky_fetch(attempt):
    """Fails until the fourth attempt."""
    if attempt < 4:
        raise ConnectionError(f"reset on attempt {attempt}")
    return ["row1", "row2"]

wait = 1
attempt = 0
total_wait = 0
data = None

while attempt < 5:
    attempt += 1
    try:
        data = flaky_fetch(attempt)
        print(f"Success on attempt {attempt}")
        break
    except ConnectionError as e:
        print(f"  {e}. waiting {wait}s")
        total_wait = total_wait + wait
        wait = wait * 2

print(f"total waited: {total_wait}s")
print(f"data: {data}")
```

```python
# DE Task 6: make a function defensive
def sum_by(rows, group_column, value_column):
    """Add up value_column per group_column. Raises on bad input."""
    if not isinstance(rows, list):
        raise TypeError(f"rows must be a list, got {type(rows).__name__}")
    if len(rows) == 0:
        raise ValueError("cannot summarise an empty table")

    totals = {}
    for row in rows:
        if group_column not in row:
            raise KeyError(f"row {row} has no column {group_column!r}")
        key = row[group_column]
        totals[key] = totals.get(key, 0) + row[value_column]
    return totals

print(sum_by([{"c": "NGN", "a": 10}, {"c": "NGN", "a": 5}], "c", "a"))

for bad_input in ["not a list", []]:
    try:
        sum_by(bad_input, "c", "a")
    except (TypeError, ValueError) as e:
        print(f"REFUSED: {type(e).__name__}: {e}")
```

## Mini project: full worked solution

```python
raw_payments = [
    {"id": 1, "user": "ahmed",  "amount": "5,000", "currency": "NGN", "status": "success"},
    {"id": 2, "user": "lahla",  "amount": "0",     "currency": "NGN", "status": "failed"},
    {"id": 3, "user": "judith", "amount": "120.50","currency": "USD", "status": "success"},
    {"id": 4, "user": "",       "amount": "800",   "currency": "NGN", "status": "success"},
    {"id": 5, "user": "uche",   "amount": "abc",   "currency": "USD", "status": "success"},
    {"id": 6, "user": "ahmed",  "amount": None,    "currency": "NGN", "status": "pending"},
    {"id": 7, "user": "judith", "amount": "300",   "currency": "NGN", "status": "success"},
    {"id": 8, "user": "lahla",  "amount": "90",                       "status": "success"},
    {"id": 3, "user": "judith", "amount": "120.50","currency": "USD", "status": "success"},
]


def parse_amount(value):
    """Return the amount as a float, or raise ValueError with a clear message."""
    if value is None:
        raise ValueError("amount is null")
    text = str(value).strip().replace(",", "")
    if text == "":
        raise ValueError("amount is empty")
    try:
        return float(text)
    except ValueError:
        raise ValueError(f"amount {value!r} is not a number")


def validate_row(row):
    """Return (True, "") for a good row, or (False, reason) for a bad one."""
    if not row.get("user"):
        return False, "empty or missing user"
    if "currency" not in row:
        return False, "missing currency column"
    try:
        parse_amount(row.get("amount"))
    except ValueError as e:
        return False, str(e)
    return True, ""


print(parse_amount(" 1,250 "))
print(validate_row(raw_payments[0]))
print(validate_row(raw_payments[4]))
```

```python
def clean_table(rows, strict=False):
    """Split rows into (clean, rejected), dropping duplicate ids."""
    clean = []
    rejected = []
    seen_ids = set()

    for row in rows:
        row_id = row.get("id")

        if row_id in seen_ids:
            rejected.append({"id": row_id, "reason": "duplicate id"})
            continue

        ok, reason = validate_row(row)
        if not ok:
            if strict:
                raise ValueError(f"row {row_id} failed validation: {reason}")
            rejected.append({"id": row_id, "reason": reason})
            continue

        seen_ids.add(row_id)
        record = dict(row)
        record["amount"] = parse_amount(row["amount"])
        clean.append(record)

    return clean, rejected


clean, rejected = clean_table(raw_payments)
print(f"clean={len(clean)} rejected={len(rejected)}")
```

```python
def count_by(rows, column):
    """Count how many rows have each distinct value of a column."""
    counts = {}
    for row in rows:
        key = row.get(column)
        counts[key] = counts.get(key, 0) + 1
    return counts


def sum_by(rows, group_column, value_column):
    """Add up value_column for each distinct value of group_column."""
    totals = {}
    for row in rows:
        key = row[group_column]
        totals[key] = totals.get(key, 0) + row[value_column]
    return totals


def top_spender(rows):
    """Return (user, total) for the biggest spender."""
    spend = sum_by(rows, "user", "amount")
    best_user = None
    best_total = 0
    for user, total in spend.items():
        if best_user is None or total > best_total:
            best_user = user
            best_total = total
    return best_user, best_total


print(count_by(clean, "status"))
print(sum_by(clean, "currency", "amount"))
print(top_spender(clean))
```

```python
def report(raw, clean, rejected):
    """Print a human readable summary of one pipeline run."""
    print("=== PAYMENTS PIPELINE REPORT ===")
    print(f"Rows in:        {len(raw)}")
    print(f"Rows loaded:    {len(clean)}")
    print(f"Rows rejected:  {len(rejected)}")
    for bad in rejected:
        print(f"   -> id {bad['id']}: {bad['reason']}")

    print()
    print(f"Status counts:  {count_by(clean, 'status')}")
    print("Totals by currency:")
    for currency, total in sum_by(clean, "currency", "amount").items():
        print(f"   {currency}: {total}")

    users = set()
    for row in clean:
        users.add(row["user"])
    print(f"Distinct users: {len(users)} {sorted(users)}")

    user, total = top_spender(clean)
    print(f"Top spender:    {user} ({total})")

    rate = len(rejected) / len(raw) * 100
    print(f"Rejection rate: {round(rate, 1)}%")


def run(rows):
    """Run the whole pipeline and return the clean rows."""
    clean, rejected = clean_table(rows)
    report(rows, clean, rejected)
    return clean


final_rows = run(raw_payments)
```
