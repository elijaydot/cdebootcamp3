#### 5. `*args` and `**kwargs` : 
- `*args` allows a function to accept any number of positional arguments.
- `**kwargs` allows a function to accept any number of keyword arguments.

difference
- `*args` collects additional positional arguments as a tuple.

The `*` has two jobs, and they are mirror images:

- In the **definition** `def f(*args)`: gather many arguments into one tuple.
- At the **call** `f(*mylist)`: spread one list out into many arguments.

Same for `**` with dictionaries and keyword arguments.

### The single most important idea: `return` vs `print`

If you only take one thing from today, take this one. `print` shows something to
a human. `return` hands a value back to your code so you can keep working with it.

Beginners mix them up constantly, and the bug is silent.

````python
# This function PRINTS. It gives nothing back.
def add_and_print(a, b):
    print(a + b)

# This function RETURNS. It hands the answer back.
def add_and_return(a, b):
    return a + b
````

````python
x = add_and_print(3, 4)      # you will see 7 on screen
````

````python
x
````

````python
y = add_and_return(3, 4)     # you will see nothing on screen
````

````python
y
````

`add_and_print` returned `None`, because a function with no `return` always
returns `None`. So `x * 2` would crash with a `TypeError`.

**Rule:** if anything else in your program needs the answer, `return` it.
Only `print` when a human is the final destination.

````python
# Proof of why it matters
def double_print(n):
    print(n * 2)

def double_return(n):
    return n * 2

numbers = [1, 2, 3]
results = []

for n in numbers:
    results.append(double_return(n))    # works

print(results)

# Try this instead and look at the result:
bad_results = []
for n in numbers:
    bad_results.append(double_print(n))

print(bad_results)      # [None, None, None]
````

````python
# return also STOPS the function immediately.
# Nothing after a return ever runs, same as break in a loop.

def check_age(age):
    if age < 0:
        return "Invalid age"       # function ends right here
    if age < 18:
        return "Minor"
    return "Adult"                 # no else needed

print(check_age(-5))
print(check_age(12))
print(check_age(30))
````

### Arguments: positional, keyword, default

- **Positional**: matched by order. `add(3, 4)` means `a=3, b=4`.
- **Keyword**: matched by name. `add(b=4, a=3)` works and is clearer.
- **Default**: has a fallback value, so the caller can leave it out.

````python
# Returning more than one value: Python packs them into a tuple

def describe(numbers):
    total = 0
    biggest = numbers[0]
    for n in numbers:
        total = total + n
        if n > biggest:
            biggest = n
    average = total / len(numbers)
    return total, average, biggest        # a tuple

result = describe([4, 9, 15, 22, 7])
print(result)
print(type(result))

# Unpack it into three names
total, average, biggest = describe([4, 9, 15, 22, 7])
print(f"total={total}, average={average}, biggest={biggest}")
````

### Real life: functions in a data pipeline

Last class you wrote loops that validated rows, counted nulls and totalled by
currency. Those were fine, but they only worked on the one table sitting in that cell.

A function turns that same logic into something you can point at **any** table.
That is the whole job: write it once, run it on everything.

````python
# Our table again
transactions = [
    {"id": 1, "user": "ahmed",  "amount": 5000,  "currency": "NGN", "status": "success"},
    {"id": 2, "user": "lahla",  "amount": 0,     "currency": "NGN", "status": "failed"},
    {"id": 3, "user": "judith", "amount": 120,   "currency": "USD", "status": "success"},
    {"id": 4, "user": "",       "amount": 800,   "currency": "NGN", "status": "success"},
    {"id": 5, "user": "uche",   "amount": -40,   "currency": "USD", "status": "success"},
]
````

````python
# DE example 1: one row in, a verdict out

def validate_row(row):
    """Return (True, "") if the row is good, or (False, reason) if it is not."""
    if row["user"] == "":
        return False, "empty user"
    if row["amount"] <= 0:
        return False, f"amount is {row['amount']}"
    return True, ""


for tx in transactions:
    ok, reason = validate_row(tx)
    if ok:
        print(f"id {tx['id']}: OK")
    else:
        print(f"id {tx['id']}: REJECTED, {reason}")
````

Returning a pair like `(True, "")` or `(False, "reason")` is a very common shape.
The caller gets both the verdict and the explanation in one call.

````python
# DE example 2: the whole validation step as one function

def split_clean_and_rejected(rows):
    """Split a table into (clean_rows, rejected_rows_with_reasons)."""
    clean = []
    rejected = []
    for row in rows:
        ok, reason = validate_row(row)
        if ok:
            clean.append(row)
        else:
            rejected.append({"row": row, "reason": reason})
    return clean, rejected


clean, rejected = split_clean_and_rejected(transactions)
print(f"clean: {len(clean)}, rejected: {len(rejected)}")
for bad in rejected:
    print(f"  id {bad['row']['id']}: {bad['reason']}")
````

````python
# DE example 3: a generic group-by that works on ANY table and ANY two columns

def sum_by(rows, group_column, value_column):
    """Add up value_column for each distinct value of group_column."""
    totals = {}
    for row in rows:
        key = row[group_column]
        if key in totals:
            totals[key] = totals[key] + row[value_column]
        else:
            totals[key] = row[value_column]
    return totals


print(sum_by(clean, "currency", "amount"))
print(sum_by(clean, "user", "amount"))
print(sum_by(clean, "status", "amount"))
````

Look at what just happened. One function, three completely different answers,
zero copied code. That is why we bother with functions.

````python
# DE example 4: counting nulls, as a reusable check

def count_nulls(rows, columns):
    """Return a dict of column -> number of null or empty values."""
    result = {}
    for column in columns:
        nulls = 0
        for row in rows:
            if row.get(column) is None or row.get(column) == "":
                nulls += 1
        result[column] = nulls
    return result


rows = [
    {"id": 1,    "email": "a@x.com", "country": "NG"},
    {"id": 2,    "email": "",        "country": "NG"},
    {"id": None, "email": "c@x.com", "country": ""},
]

print(count_nulls(rows, ["id", "email", "country"]))
````

`row.get(column)` is a dict method that returns `None` instead of crashing when the
key is missing. `row["missing"]` raises `KeyError`, `row.get("missing")` returns `None`.
In pipelines, `.get()` is usually what you want.

````python
# DE example 5: generating SQL, now parameterised

def null_check_sql(table, columns, schema="public"):
    """Return a list of null-check statements for one table."""
    statements = []
    for column in columns:
        statements.append(
            f"SELECT '{table}' AS tbl, '{column}' AS col, COUNT(*) AS nulls "
            f"FROM {schema}.{table} WHERE {column} IS NULL;"
        )
    return statements


for sql in null_check_sql("customers", ["id", "email"], schema="raw"):
    print(sql)
````

````python
# DE example 6: a small pipeline built from the functions above

def run_pipeline(rows, name="unnamed"):
    """Validate, summarise and report on a table. Returns the clean rows."""
    clean, rejected = split_clean_and_rejected(rows)
    totals = sum_by(clean, "currency", "amount")

    print(f"--- {name} ---")
    print(f"in: {len(rows)}, clean: {len(clean)}, rejected: {len(rejected)}")
    for currency, total in totals.items():
        print(f"  {currency}: {total}")
    return clean


good_rows = run_pipeline(transactions, name="transactions")
````


### The distinction worth keeping:

## Function, Method, Attribute

**Function**: standalone, called on its own.  
`len(raw)`, `print(raw)`, `int(raw)`

**Method**: belongs to an object, reached with a dot, and you can run it.  
`raw.isdigit()`, `raw.upper()`, `raw.strip()`

**Attribute**: anything you reach with a dot.  
A method is just an attribute you can run.

So `len()` is a function that takes a string. `.upper()` is a method of a string. Different shapes.

### The dot and the parentheses do different jobs

The dot fetches the thing. The parentheses run it.

```python
raw = "  25  "

raw.strip     # <built-in method strip of str object>   fetched, not run
raw.strip()   # "25"                                    now it ran
```

`raw.strip` is still a method. It just hasn't been called yet.

**Common bug:** `if raw.isdigit:` (no parens) is always True, because a method
object is truthy no matter what the string holds. You need `raw.isdigit()`.

### Two kinds of attributes

Some give you a result straight away. Some give you a function to run.

```python
import pandas as pd

df.shape      # (1000, 5)                    result -> data attribute, no parens
df.head       # <bound method ... head>      function -> method, needs parens
df.head()     # actual rows
```

Quick check: if the output starts with `<bound method` or `<built-in method`,
you forgot the parentheses.

| Term | What it means |
|---|---|
| Attribute | Anything after a dot. The umbrella word. |
| Data attribute | An attribute that gives a value. No parentheses. |
| Method | An attribute you can run. Needs parentheses. |
| Function | Standalone, no dot, no object. |

### Strings are all methods

A string has 47 public attributes and every single one is a method.
On a string, always use the parentheses.

```python
s = "hello"
s.upper()            # "HELLO"
s.count("l")         # 2
s.replace("h", "j")  # "jello"
```

Integers are mixed, which shows both kinds side by side:

```python
n = 25
n.numerator      # 25   data attribute, no parens
n.bit_length()   # 5    method, needs parens
```

### Why `num.isdigit()` fails

```python
num = int(input("give ur age: "))
num.isdigit()   # AttributeError: 'int' object has no attribute 'isdigit'
```

`isdigit` lives on `str`, not `int`. Once `int()` converts it, the string
methods are gone. `AttributeError` literally means "this object has nothing
by that name".

Check the string first, then convert:

```python
raw = input("give ur age: ")
if raw.isdigit():
    num = int(raw)
else:
    print("that's not a number")
```

### Rule of thumb

Everything after a dot is an **attribute**. The ones you can run are called
**methods**. No dot, standing alone, is a **function**.

### Writing your own module, for real

A module is nothing more than a `.py` file sitting next to your notebook.
The cell below creates one. `%%writefile` is a Jupyter command that saves
everything under it to a file. It must be the very first line of the cell.

````python
%%writefile de_utils.py
"""Small helpers for the DEC data pipeline class."""


def clean_amount(value):
    """Turn a messy amount into a float. Returns 0.0 when it cannot be parsed."""
    text = str(value).strip().replace(",", "")
    if text == "" or text.lower() == "none":
        return 0.0
    return float(text)


def is_valid_email(email):
    """True when the string looks like an email address."""
    if "@" not in email:
        return False
    name, _, domain = email.partition("@")
    return name != "" and "." in domain


def count_by(rows, column):
    """Count how many rows have each distinct value of a column."""
    counts = {}
    for row in rows:
        key = row.get(column)
        counts[key] = counts.get(key, 0) + 1
    return counts


def sum_by(rows, group_column, value_column):
    """Add up value_column for each distinct value of group_column."""
    totals = {}
    for row in rows:
        key = row[group_column]
        totals[key] = totals.get(key, 0) + row[value_column]
    return totals


if __name__ == "__main__":
    # This only runs when you execute the file directly:  python de_utils.py
    # It does NOT run when someone imports the module.
    print("Running de_utils.py directly as a script")
    print(clean_amount(" 1,250 "))
    print(is_valid_email("ahmed@dataencode.com"))
````

````python
# Now use it
import de_utils

print(de_utils.clean_amount(" 12,500 "))
print(de_utils.is_valid_email("ahmed@dataencode.com"))
print(de_utils.is_valid_email("nope"))
````

````python
# Or bring in just what you need
from de_utils import count_by, sum_by

transactions = [
    {"id": 1, "user": "ahmed",  "amount": 5000, "currency": "NGN", "status": "success"},
    {"id": 2, "user": "lahla",  "amount": 900,  "currency": "NGN", "status": "failed"},
    {"id": 3, "user": "judith", "amount": 120,  "currency": "USD", "status": "success"},
]

print(count_by(transactions, "status"))
print(sum_by(transactions, "currency", "amount"))
````

### Handling Exceptions
Python provides a way to handle exceptions using the try, except, finally, and else blocks.

- `try` block: Code that might raise an exception is placed here.
- `except` block: Code that runs if an exception is raised.
- `else` block: Code that runs if no exception is raised.
- `finally` block: Code that runs no matter what, even if an exception was raised.

#### Example:

### Error hierarchy (simplified)

Every exception in Python is part of a family tree. Catching a parent catches all
of its children.

```
BaseException
 └── Exception
      └── OSError
           └── IOError
                └── FileNotFoundError
```

So `except OSError` also catches `FileNotFoundError`, and `except Exception` catches
almost everything. That is exactly why the order of your `except` blocks matters.

### The five you will meet most often

- `ValueError` — right type, impossible value. `int("hello")`
- `TypeError` — wrong type entirely. `5 + "string"`
- `ZeroDivisionError` — divided by zero
- `KeyError` — that dict key is not there
- `IndexError` — that list position is not there

### Why `except Exception as e` goes last

`Exception` is the parent of nearly every error, so it matches nearly everything.

Put it at the **bottom** and it works as a safety net: the specific handlers above
it get first refusal, and anything you did not anticipate still gets caught.

Put it at the **top** and it swallows everything. Every specific handler below it
becomes dead code that never runs.

````python
try:
    x = int(input("Enter a number: "))
    result = 10 / x
    print(f"Result is {result}")
except ValueError:
    print("Please enter a valid integer.")
except ZeroDivisionError:
    print("Division by zero is not allowed.")
except exception as e:
    print("thank you for surprising me")
else:
    print("No errors occurred!")
finally:
    print("This will always execute.")
````

### `try` / `except` / `else` / `finally`

- `try`: the risky code
- `except`: what to do when it goes wrong
- `else`: runs only if **nothing** went wrong
- `finally`: runs **always**, error or not. Use it for cleanup

````python
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("  cannot divide by zero")
        return None
    else:
        print("  no errors")
        return result
    finally:
        print("  finally always runs")


print("10 / 2:")
print(safe_divide(10, 2))

print()
print("10 / 0:")
print(safe_divide(10, 0))
````

### Two bugs in the try/except further up this notebook

```python
except exception as e:          # bug 1
    print("thank you for surprising me")
```

**Bug 1: lowercase `exception`.** Python's built-in is `Exception` with a capital E.
`exception` does not exist, so the moment an unexpected error arrives, this handler
itself blows up with a `NameError`. Worse, it looks fine until the day it is needed.

```python
except:                          # bug 2
    print("hey enter something valid.")
```

**Bug 2: a bare `except`.** It catches *everything*, including you pressing the stop
button (`KeyboardInterrupt`) and genuine bugs in your own code. Your pipeline then
reports "enter something valid" when the real problem was a typo on line 40.

Catch what you can actually handle. Put `except Exception as e` last, as a safety net,
and print the real error.

````python
# The corrected version
def get_number_and_divide(raw_input_value):
    try:
        x = int(raw_input_value)
        result = 10 / x
    except ValueError:
        return "Please enter a valid integer"
    except ZeroDivisionError:
        return "Division by zero is not allowed"
    except Exception as e:                       # safety net, LAST
        return f"Something unexpected: {type(e).__name__}: {e}"
    else:
        return f"Result is {result}"
    finally:
        print("  (cleanup would go here)")


print(get_number_and_divide("5"))
print(get_number_and_divide("hello"))
print(get_number_and_divide("0"))
print(get_number_and_divide(None))
````

### Order matters, and it is not obvious

Python checks your `except` blocks **top to bottom** and uses the **first** one that
matches. `Exception` is the parent of nearly everything, so if you put it first,
nothing below it ever runs.

````python
# WRONG ORDER: the specific handler is unreachable
try:
    int("hello")
except Exception:
    print("generic handler caught it")
except ValueError:
    print("this line can never run")
````

````python
# RIGHT ORDER: specific first, general last
try:
    int("hello")
except ValueError:
    print("ValueError handler caught it, which is what we wanted")
except Exception:
    print("generic net, only for the unexpected")
````

````python
# Catching two exceptions with one handler
for value in ["10", "hello", "0", ""]:
    try:
        number = int(value)
        print(f"{value!r} -> 100/{number} = {100 / number}")
    except (ValueError, ZeroDivisionError) as e:
        print(f"{value!r} -> skipped, {type(e).__name__}: {e}")
````

`{value!r}` prints the value the way Python would write it, with quotes around
strings. Very handy in logs, because it shows you the difference between `0` and `"0"`
and `""`.

### Raising your own errors

`try/except` is for handling other people's problems. `raise` is for announcing yours.

Raise when your function is given something it cannot honestly work with. Failing
loudly at the source beats returning a wrong number that reaches a dashboard.

````python
def load_batch(rows, max_rows=1000):
    if not isinstance(rows, list):
        raise TypeError(f"rows must be a list, got {type(rows).__name__}")
    if len(rows) == 0:
        raise ValueError("refusing to load an empty batch")
    if len(rows) > max_rows:
        raise ValueError(f"batch too large: {len(rows)} rows, limit is {max_rows}")
    return f"Loaded {len(rows)} rows"


print(load_batch([{"id": 1}, {"id": 2}]))

for bad in ["not a list", [], [{"id": i} for i in range(1500)]]:
    try:
        print(load_batch(bad))
    except (TypeError, ValueError) as e:
        print(f"REFUSED -> {type(e).__name__}: {e}")
````

### Real life: error handling in a data pipeline

The question is never "should I handle errors". It is **"at what level?"**

- Wrap the **whole job** in a try and one bad row kills the run.
- Wrap **each row** and one bad row is quarantined while the other 99,999 load.

Row-level handling is the default in analytics pipelines. Job-level failure is right
for money movement, where a half-loaded batch is worse than nothing.

````python
# DE example 1: a conversion that can never crash your job

def to_float(value, default=0.0):
    """Convert anything to a float. Returns default when it cannot."""
    try:
        return float(str(value).strip().replace(",", ""))
    except (ValueError, TypeError):
        return default


for raw in ["1,250", " 99.5 ", "", None, "abc", 42]:
    print(f"{raw!r:>10}  ->  {to_float(raw)}")
````

````python
# DE example 2: per-row error handling with a quarantine

raw_rows = [
    {"id": 1, "user": "ahmed",  "amount": "5,000"},
    {"id": 2, "user": "lahla",  "amount": "abc"},
    {"id": 3,                   "amount": "120"},        # no "user" key at all
    {"id": 4, "user": "uche",   "amount": None},
    {"id": 5, "user": "judith", "amount": "300"},
]

loaded = []
quarantine = []

for row in raw_rows:
    try:
        clean = {
            "id": row["id"],
            "user": row["user"],                 # KeyError if missing
            "amount": float(row["amount"]),      # ValueError or TypeError if bad
        }
        loaded.append(clean)
    except KeyError as e:
        quarantine.append({"row": row, "error": f"missing column {e}"})
    except (ValueError, TypeError) as e:
        quarantine.append({"row": row, "error": f"bad amount: {e}"})

print(f"loaded: {len(loaded)}, quarantined: {len(quarantine)}")
print()
for bad in quarantine:
    print(f"  id {bad['row']['id']}: {bad['error']}")
````

Row 1 also failed, quietly: `float("5,000")` cannot parse the comma. Look at the
output and check whether it made it in. This is exactly why you print the quarantine
instead of trusting the count.

````python
# DE example 3: retry with try/except inside a loop
# The pattern behind every "connection reset by peer" you will ever see.

def flaky_fetch(attempt):
    """Pretend this is an API. It fails the first two times."""
    if attempt < 3:
        raise ConnectionError(f"connection reset (attempt {attempt})")
    return ["row1", "row2", "row3"]


max_retries = 5
attempt = 0
data = None

while attempt < max_retries:
    attempt += 1
    try:
        data = flaky_fetch(attempt)
        print(f"Success on attempt {attempt}")
        break
    except ConnectionError as e:
        print(f"  attempt {attempt} failed: {e}")

if data is None:
    print("All retries exhausted. Failing the job")
else:
    print(f"Got {len(data)} rows")
````

````python
# DE example 4: finally, for cleanup that must happen either way

def load_to_warehouse(rows):
    connection = "OPEN"
    print(f"connection is {connection}")
    try:
        if len(rows) == 0:
            raise ValueError("no rows to load")
        print(f"loading {len(rows)} rows")
        return len(rows)
    except ValueError as e:
        print(f"load failed: {e}")
        return 0
    finally:
        connection = "CLOSED"
        print(f"connection is {connection}")


print(load_to_warehouse([{"id": 1}]))
print()
print(load_to_warehouse([]))
````

Without `finally`, the connection stays open when the error path is taken. Do that
a few thousand times and the warehouse starts refusing new connections. This is a real
outage, not a theoretical one.

````python
# DE example 5: the anti-pattern, so you can recognise it in a code review

rows = [{"id": 1, "amount": "abc"}, {"id": 2, "amount": "50"}]

# BAD: swallows the error and tells nobody
for row in rows:
    try:
        amount = float(row["amount"])
    except:
        pass          # the row silently disappears. nobody ever knows

print("Finished. (And you have no idea a row went missing.)")
print()

# GOOD: handle it, and say something
skipped = 0
for row in rows:
    try:
        amount = float(row["amount"])
    except (ValueError, TypeError) as e:
        skipped += 1
        print(f"WARN row {row['id']}: {type(e).__name__}: {e}")

print(f"Finished. {skipped} row(s) skipped out of {len(rows)}")
````

> **Errors should never pass silently. Unless explicitly silenced.**
>
> That is straight from the Zen of Python you printed last class. `except: pass` is
> the most expensive two words in data engineering.

## Common mistakes to watch for

### Functions

| Mistake | What you see | The fix |
|---|---|---|
| `print` instead of `return` | The value is `None` when you use it later | `return` it. `print` only for humans |
| Defining but never calling | Nothing happens at all | `def greet():` defines it, `greet()` runs it |
| `&&`, `\|\|`, `!` | `SyntaxError` | `and`, `or`, `not` |
| Default before non-default | `SyntaxError` | `def f(a, b=1)`, never `def f(a=1, b)` |
| Mutable default `def f(x, batch=[])` | Old data appears in new calls | Use `batch=None` then `if batch is None: batch = []` |
| Expecting an inside variable outside | `NameError` | `return` it out |
| `collect(mylist)` when you meant many args | One tuple holding one list | `collect(*mylist)` |

### Modules

| Mistake | What you see | The fix |
|---|---|---|
| `from x import a` then calling `b` | `NameError: name 'b' is not defined` | Import `b` too, or `import x` and use `x.b` |
| Edited the .py, change not showing | Old behaviour | Restart the kernel, or `importlib.reload(module)` |
| `from module import *` | Mysterious name clashes | Import what you need, by name |
| Module file not next to the notebook | `ModuleNotFoundError` | Same folder, or add it to the path |

### Errors

| Mistake | What you see | The fix |
|---|---|---|
| `except exception` | `NameError` at the worst moment | Capital E: `except Exception as e` |
| Bare `except:` | Real bugs disguised as user errors | Name the exceptions you expect |
| `except: pass` | Rows vanish, nobody knows | At minimum print or log it |
| `except Exception` first | Specific handlers never run | Specific first, general last |
| Trying to catch a `SyntaxError` | It fires before `try` exists | Just fix the syntax |
| Wrapping the whole job | One bad row kills 100k good ones | Wrap each row, quarantine the bad ones |
