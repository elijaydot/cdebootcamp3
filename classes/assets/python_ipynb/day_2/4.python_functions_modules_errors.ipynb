####

#### 2. `Generic Functions` : 
A generic function can take parameters (arguments) to make it more versatile.

````python
# The greet function now takes a parameter name and uses it to customize the greeting.
def greet(name):
    print(f"Hello, {name}!")

greet("Alice")  # Output: Hello, Alice!
greet("Bob")    # Output: Hello, Bob!
````

###

#### 3. `Non-Default Arguments` : 
Non-default arguments are required parameters that must be provided when calling a function.

````python
# add is a function that takes two non-default arguments, a and b, and returns their sum.
def add(a, b):
    return a + b

result = add(3, 4)  # Output: 7
````

````python
result
````

###

#### 4.  `Default Arguments` : 
Default arguments are parameters that assume a default value if no value is provided when the function is called.

````python
#The greet function has a default argument greeting with a default value of "Hello". If no greeting is provided, it uses the default.
def greet(name, greeting="Hello"):
    print(f"{greeting}, {name}!")

greet("Alice")          # Output: Hello, Alice!
greet("Bob", "Hi")      # Output: Hi, Bob!
````

###

#### 5. `*args` and `**kwargs` : 
- `*args` allows a function to accept any number of positional arguments.
- `**kwargs` allows a function to accept any number of keyword arguments.

difference
- `*args` collects additional positional arguments as a tuple.
A positional argument is an argument that is passed to a function based on its position or order in the function call. The function assigns each positional argument to the corresponding parameter in the function definition, starting from the first parameter. Its like list and/or tuple

- `**kwargs` collects additional keyword arguments as a dictionary.




````python
def collect_details(*args):
    print(type(args))
    for i in args:
        print(i)
````

````python
details= ['enquiry', 'Hammed', 'dataguy', 200, 500, 'Blessing']
````

````python
collect_details(details)
````

````python
def print_args(*args):
    for arg in args:
        print(arg)

print_args(1, 2, 3)  # Output: 1 2 3
````

````python
def print_kwargs(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_kwargs(name="Alice", age=25, location="lagos")
````

````python
def print_people(people):
    for person in people:
        for key, value in person.items():
            print(f"{key}: {value}")
        print()  # Print a newline for better readability

people = [
    {"name": "Alice", "age": 25, "location": "Lagos"},
    {"name": "Bob", "age": 30, "location": "New York"},
    {"name": "Charlie", "age": 35, "location": "London"}
]

print_people(people)
````


### The distinction worth keeping:

## Function, Method, Attribute

**Function**: standalone, called on its own.  
`len(raw)`, `print(raw)`, `int(raw)`

**Method**: belongs to an object, reached with a dot, and you can run it.  
`raw.isdigit()`, `raw.upper()`, `raw.strip()`

**Attribute**: anything you reach with a dot.  
A method is just an attribute you can run.

So `len()` is a function that takes a string. `.upper()` is a method of a string. Different shapes.

### The dot and the parentheses do different jobs

The dot fetches the thing. The parentheses run it.

```python
raw = "  25  "

raw.strip     # <built-in method strip of str object>   fetched, not run
raw.strip()   # "25"                                    now it ran
```

`raw.strip` is still a method. It just hasn't been called yet.

**Common bug:** `if raw.isdigit:` (no parens) is always True, because a method
object is truthy no matter what the string holds. You need `raw.isdigit()`.

### Two kinds of attributes

Some give you a result straight away. Some give you a function to run.

```python
import pandas as pd

df.shape      # (1000, 5)                    result -> data attribute, no parens
df.head       # <bound method ... head>      function -> method, needs parens
df.head()     # actual rows
```

Quick check: if the output starts with `<bound method` or `<built-in method`,
you forgot the parentheses.

| Term | What it means |
|---|---|
| Attribute | Anything after a dot. The umbrella word. |
| Data attribute | An attribute that gives a value. No parentheses. |
| Method | An attribute you can run. Needs parentheses. |
| Function | Standalone, no dot, no object. |

### Strings are all methods

A string has 47 public attributes and every single one is a method.
On a string, always use the parentheses.

```python
s = "hello"
s.upper()            # "HELLO"
s.count("l")         # 2
s.replace("h", "j")  # "jello"
```

Integers are mixed, which shows both kinds side by side:

```python
n = 25
n.numerator      # 25   data attribute, no parens
n.bit_length()   # 5    method, needs parens
```

### Why `num.isdigit()` fails

```python
num = int(input("give ur age: "))
num.isdigit()   # AttributeError: 'int' object has no attribute 'isdigit'
```

`isdigit` lives on `str`, not `int`. Once `int()` converts it, the string
methods are gone. `AttributeError` literally means "this object has nothing
by that name".

Check the string first, then convert:

```python
raw = input("give ur age: ")
if raw.isdigit():
    num = int(raw)
else:
    print("that's not a number")
```

### Rule of thumb

Everything after a dot is an **attribute**. The ones you can run are called
**methods**. No dot, standing alone, is a **function**.

Errors and exceptions in Python are events that disrupt the normal flow of a program's execution. Errors are the problems in the program that cause it to crash, while exceptions are the issues that a program encounters while it is running but can be handled.

### Handling Exceptions
Python provides a way to handle exceptions using the try, except, finally, and else blocks.

- `try` block: Code that might raise an exception is placed here.
- `except` block: Code that runs if an exception is raised.
- `else` block: Code that runs if no exception is raised.
- `finally` block: Code that runs no matter what, even if an exception was raised.

#### Example:

````python
Error Hierarchy (Simplified)
BaseException
 └── Exception
      └── OSError
           └── IOError
                └── FileNotFoundError
````

````python
ValueError
TypeError
ZeroDivisionError
KeyError
IndexError
````

````python
Advanced Tip (Since You're Progressing Well)

This is why people often write:

except Exception as e:

at the bottom — because Exception catches almost everything.

But if you put it first, it blocks all specific ones.
````

````python
try:
    x = int(input("Enter a number: "))
    result = 10 / x
    print(f"Result is {result}")
except ValueError:
    print("Please enter a valid integer.")
except ZeroDivisionError:
    print("Division by zero is not allowed.")
except Exception as e:
    print("thank you for surprising me")
else:
    print("No errors occurred!")
finally:
    print("This will always execute.")
````
